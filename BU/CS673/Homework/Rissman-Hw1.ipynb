```python
# Q1 Multiplication table
# Write a program that prints a multiplication table for numbers up to 12.

n = 12

for i in range(1,n+1):
    for j in range (1,n+1):
        print(f"{i} x {j} = {i * j}", end="\t")
    print()
```

```python
# Q2 Palindrome
# Write a program that asks a user to input a string and then determine if it’s a palindrome.
# If it is, print out “The string  [is | is not] a palindrome. If the user types q, the program exits.

from ipywidgets import widgets, Layout

def is_palindrome(string):
  """
  Checks if the input string is a palindrome.

  Args:
    input_string: The string to be checked.

  Returns:
    True if the string is a palindrome, False otherwise.
  """
  input = string.replace(" ", "").strip().lower()
  return input == input[::-1]

# Create input and output widgets
input_text = widgets.Text(
  description="Enter a string:",
  layout=Layout(width='auto'),
  style={'description_width': 'initial'},
  continuous_update=False
  )
output_text = widgets.Output()
with output_text:
  print("Type in a string then press enter to check if it is a palindrome.\nEnter 'q' to quit.")

# Define the function to update the output when the input changes
def on_text_change(change):
  with output_text:
    output_text.clear_output()
    string = change['new']
    if string == 'q':
      input_text.disabled=True
      print("Program exited")
    elif is_palindrome(string):
      print("The string is a palindrome.\nEnter 'q' to quit.")
    else:
      print("The string is not a palindrome.\nEnter 'q' to quit.")


# Attach the event handler to the input widget
input_text.observe(on_text_change, names='value')

# Display the widgets
display(input_text, output_text)
```

```python
# Q3 Combine strings
# Write a program that asks the user to input two lists of equal length in the format:
# [a,b,c], [1,2,3]. The program should test that both lists are of equal length, if not, print out an error message.
# If they are the same length, combine the two lists by alternatingly taking elements and output the results.
# Using the input example above, the output would be: [a,1,b,2,c,3].

import ast

def combine_lists(list1_str, list2_str):
  """
  Combines two lists of equal length by alternatingly taking elements.

  Args:
    list1: The first list.
    list2: The second list.

  Returns:
    A new list containing the combined elements.

  Raises:
    ValueError: If the lists have different lengths or invalid format.
  """
  # Attempt to convert from string to list
  try:
    list1 = ast.literal_eval(list1_str)
    list2 = ast.literal_eval(list2_str)
  except (SyntaxError, ValueError):
    raise ValueError("Invalid list format. Please enter lists in the format: ['a', 'b', 'c'] or [1, 2, 3]")

  # Check if both inputs are lists
  if not isinstance(list1, list) or not isinstance(list2, list):
    raise ValueError("Invalid input. Please enter valid lists.")

  # Check if both lists have the same length
  if len(list1) != len(list2):
    raise ValueError("Lists must have the same length.")

  combined_list = []
  for i in range(len(list1)):
    combined_list.append(list1[i])
    combined_list.append(list2[i])
  return combined_list

# Create input widgets
list1_input = widgets.Text(description="List 1:")
list2_input = widgets.Text(description="List 2:")
output_widget = widgets.Output()

# Define the function to be called when the button is clicked
def on_button_click(b):
  try:
    combined_list = combine_lists(list1_input.value, list2_input.value)
    with output_widget:
      output_widget.clear_output()
      print("Combined List: ", combined_list)
  except ValueError as e:
    with output_widget:
      output_widget.clear_output()
      print("Error: ", str(e))
  except Exception as e:
    with output_widget:
      output_widget.clear_output()
      print("An unexpected error occurred: ", str(e))

# Create a button to trigger the function
button = widgets.Button(description="Combine")
button.on_click(on_button_click)

# Display the widgets
display(list1_input, list2_input, button, output_widget)
```

```python
# Q5 Is it a leap year
# Write a program that determines if an inputted year is a leap year.

def is_leap_year(year):
  try:
    year = int(year)
  except ValueError:
    raise ValueError("Invalid year format. Please enter the year: YYYY")
  if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    return True
  else:
    return False
  
# Create input and output widgets
input_text = widgets.IntText(
  description="Enter year:",
  style={'description_width': 'initial'},
  continuous_update=False
  )
output_text = widgets.Output()


# Define the function to update the output when the input changes
def on_text_change(change):
  try:
    with output_text:
      output_text.clear_output()
      year = change['new']
      if is_leap_year(year):
        print(f"{year} is a leap year!")
      else:
        print(f"{year} is not a leap year.")
  except ValueError as e:
    with output_widget:
      output_widget.clear_output()
      print("Error: ", str(e))
  except Exception as e:
    with output_widget:
      output_widget.clear_output()
      print("An unexpected error occurred: ", str(e))



# Attach the event handler to the input widget
input_text.observe(on_text_change, names='value')

# Display the widgets
display(input_text, output_text)
```
