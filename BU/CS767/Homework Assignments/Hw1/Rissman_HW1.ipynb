```python
# Write a Python function extreme_args() that takes a list as input and returns its argmin and argmax.
# It’s impossible to compute the argmin and argmax without a function to evaluate the list of inputs.
# The function extreme_args() will need to take both a function and the list of inputs. 

def extreme_args(function, inputs):
    """
    Input: A function and a list of inputs. The function must take one argument and return a numerical value.

    Outputs: A tuple containing two lists: the arguments that minimize the function's output
             and the arguments that maximize the function's output.  If multiple inputs
             produce the same minimum or maximum value, all are included in the respective list.
    """
    # Initialize argmin and argmax with the first input and its corresponding function value.
    argmin = [function(inputs[0])]
    argmax = [function(inputs[0])]

    for i in inputs:
        output_value = function(i)
        output_min = function(argmin[0])
        output_max = function(argmax[0])

        if output_value < output_min:
            argmin = [i]
        elif output_value == output_min:
            argmin.append(i)
        elif output_value > output_max:
            argmax = [i]
        elif output_value == output_max:
            argmax.append(i)
    
    return argmin, argmax


xs = [-3.5, 5, -10, 0, 15.5, -2]
def square(x):
    return x * x

print(extreme_args(square, xs))
```
