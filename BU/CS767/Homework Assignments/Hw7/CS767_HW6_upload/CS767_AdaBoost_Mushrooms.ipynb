# Edible or Poisonous?

* For this exercise, we will use a dataset from the **UC Irvine Machine Learning Repository**.
* This data set includes descriptions of hypothetical samples corresponding to 23 species of gilled mushrooms in the Agaricus and Lepiota Family (Data Source: https://archive.ics.uci.edu/ml/datasets/Mushroom, pp. 500-525).
* Each species is identified as `definitely edible`, `definitely poisonous`, or of unknown edibility and not recommended. This latter class was combined with the poisonous one.
* The Guide clearly states that there is no simple rule for determining the edibility of a mushroom; no rule like [``leaflets three, let it be'' for Poisonous Oak and Ivy](https://purduelandscapereport.org/article/leaves-of-three-let-it-be-should-be-leaflets-of-three/).
