# Training and Evaluation
* Create $\mathcal{D}_{train}$ with 85% of the datapoints (use a random state = 101).
* Do you need to use 1-hot encoding?

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
X = df.drop('class', axis=1)
y = df['class']
X = encoder.fit_transform(X, y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=101)
print(f"Training set size: {len(X_train)} ({len(X_train)/len(df):.2%} of the data)")
print(f"Testing set size: {len(X_test)} ({len(X_test)/len(df):.2%} of the data)")
```

Yes, one-hot encoding is necessary because all features are categorical.
One-hot encoding will convert categorical variables to a format suitable for machine learning models.

## Model Training
* Import the AdaBoostClassifier class and train an object of the class with a single estimator.
* What is the test accuracy?
* Plot the accuracy of the boosted learner as the number of weak learners increases from 1 to 50 (the figure shows a few more).
* Show the first five weak learners in the 20th boosted tree.
  * What are such small trees called in the literature?

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
# Create and train an AdaBoost classifier with a single estimator
ada = AdaBoostClassifier(n_estimators=1, random_state=101)
ada.fit(X_train, y_train)
```

```python
y_pred = ada.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Test accuracy with a single estimator: {accuracy:.4f}")
```

```python
n_estimators_range = range(1, 51)
accuracies = []

for n_estimators in n_estimators_range:
    ada = AdaBoostClassifier(n_estimators=n_estimators, random_state=101)
    ada.fit(X_train, y_train)
    y_pred = ada.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)

# Plot the results
plt.plot(n_estimators_range, accuracies)
plt.xlabel('Weak Learners')
plt.ylabel('Accuracy')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
ada_20 = AdaBoostClassifier(n_estimators=20, random_state=101)
ada_20.fit(X_train, y_train)
```

```python
for i in range(5):
    plot_tree(ada_20.estimators_[i], filled=True)
    plt.title(f"Weak Learner {i}")
    plt.tight_layout()
    plt.show()
```

These small trees are called decision stumps in the literature.
