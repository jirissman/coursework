```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)

# Sample size
N = 10000

# Normal distribution
normal_samples = stats.norm.rvs(loc=0, scale=1, size=N)

# Exponential distribution
exponential_samples = stats.expon.rvs(scale=1, size=N)

# Poisson distribution
poisson_samples = stats.poisson.rvs(mu=5, size=N)


# Create plots
plt.figure(figsize=(5, 10))

# Normal distribution histogram
plt.subplot(3, 1, 1)
plt.hist(normal_samples, bins=50, density=True, alpha=0.7)
plt.title('Normal Distribution\nN=10000, μ=0, σ=1')
x = np.linspace(min(normal_samples), max(normal_samples), 1000)
plt.ylabel('Density')

# Exponential distribution histogram
plt.subplot(3, 1, 2)
plt.hist(exponential_samples, bins=50, density=True, alpha=0.7)
plt.title('Exponential Distribution\nN=10000, λ=1')
x = np.linspace(0, max(exponential_samples), 1000)
plt.ylabel('Density')

# Poisson distribution histogram
plt.subplot(3, 1, 3)
plt.hist(poisson_samples, bins=np.arange(0, max(poisson_samples) + 1.5) - 0.5, 
         density=True, alpha=0.7)
plt.title('Poisson Distribution\nN=10000, λ=5')
x = np.arange(0, max(poisson_samples) + 1)
plt.ylabel('Probability')

plt.tight_layout()
plt.show()
```
