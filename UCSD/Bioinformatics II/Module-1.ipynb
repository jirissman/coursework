# String Composition Problem
Generate the k-mer composition of a string.

**Input**: An integer *k* and a string *Text*.

**Output**: Composition$_k$(*Text*) (the *k*-mers can be provided in any order).

**Code Challenge**: Solve the String Composition Problem.

**Sample Input**:

```
5
CAATCCAAC
```

**Sample Output**:

```
CAATC AATCC ATCCA TCCAA CCAAC
```

````python
def GenerateComposition(k, text):
    n = len(text)
    composition = []
    for i in range(n - k + 1):
        kmer = text[i:i + k]
        composition.append(kmer)
    return composition
````

````python
# Sample Input
k = 5
text = "CAATCCAAC"
composition = GenerateComposition(k, text)
print(*composition)

# Test Assertion
expected_output = 'CAATC AATCC ATCCA TCCAA CCAAC'.split()
assert composition == expected_output, f"Expected {expected_output}, but got {composition}"
print("Test passed!")
````

# String Spelled by a Genome Path Problem
Reconstruct a string from its genome path.

**Input**: A sequence path of *k*-mers *Pattern*$_1$, … ,*Pattern*$_n$ such that the last *k* - 1 symbols of *Pattern*$_i$ are equal to the first *k*-1 symbols of *Pattern*$_{i+1}$ for 1 ≤ *i* ≤ *n*-1.

**Output**: A string *Text* of length *k*+*n*-1 such that the *i*-th *k*-mer in *Text* is equal to *Pattern*$_i$ (for 1 ≤ *i* ≤ *n*).

Reconstructing the genome from its genome path is easy: as we proceed from left to right, the 3-mers “spell’ out TAATGCCATGGGATGTT, adding one new symbol to the genome at each new 3-mer. This yields a function PathToGenome(*path*).

Unfortunately, constructing the genome path requires us to know the genome in advance.

**Code Challenge**: Solve the String Spelled by a Genome Path Problem.

**Sample Input**:

```
ACCGA CCGAA CGAAG GAAGC AAGCT
```

**Sample Output**:

```
ACCGAAGCT
```

````python
def PathToGenome(path):
    genome = path[0]
    for kmer in path[1:]:
        genome += kmer[-1]
    return genome
````

````python
# Sample Input
path = "ACCGA CCGAA CGAAG GAAGC AAGCT".split()

# Run the function
genome = PathToGenome(path)
print(genome)

# Test Assertion
expected_output = "ACCGAAGCT"
assert genome == expected_output, f"Expected {expected_output}, but got {genome}"
print("Test passed!")
````

# Overlap Graph Problem
Solve the Overlap Graph Problem (restated below).

**Input**: A collection *Patterns* of *k*-mers.

**Output**: The overlap graph Overlap(*Patterns*), in the form of an adjacency list. (You may return the nodes and their edges in any order.)

**Note**: You don't need to account for repeated elements in *Patterns* in this problem.

**Code Challenge**: Solve the Overlap Graph Problem.

**Sample Input**:

```
ATGCG GCATG CATGC AGGCA GGCAT GGCAC
```

**Sample Output**:

```
CATGC: ATGCG
GCATG: CATGC
GGCAT: GCATG
AGGCA: GGCAC GGCAT
```

````python
def Overlap(patterns):
    graph = {}
    for pattern in patterns:
        graph[pattern] = []
        suffix = pattern[1:]
        for potential_match in patterns:
            if potential_match.startswith(suffix):
                graph[pattern].append(potential_match)
    return graph
````

````python
# Sample Input
patterns = "ATGCG GCATG CATGC AGGCA GGCAT GGCAC".split()

# Run the function
adjacency_list = Overlap(patterns)

# Print the output
for pattern in sorted(adjacency_list):
    if adjacency_list[pattern]:
        print(f"{pattern}: {' '.join(sorted(adjacency_list[pattern]))}")

# Test Assertion
expected_edges = {
    'CATGC': ['ATGCG'],
    'GCATG': ['CATGC'],
    'GGCAT': ['GCATG'],
    'AGGCA': ['GGCAC', 'GGCAT']
}

# Normalize for comparison
formatted_output = {}
for k, v in adjacency_list.items():
    if v:
        formatted_output[k] = sorted(v)

expected_normalized = {k: sorted(v) for k, v in expected_edges.items()}

# Check if all expected keys are present and values match
for k, v in expected_normalized.items():
    assert k in formatted_output, f"Key {k} missing from output"
    assert formatted_output[k] == v, f"For key {k}, expected {v}, got {formatted_output[k]}"

print("Test passed!")
````

# De Bruijn Graph from a String Problem
Construct the de Bruijn graph of a string.

**Input**: An integer *k* and a string *Text*.

**Output**: DeBruijn$_k$(*Text*), in the form of an adjacency list.

**Code Challenge**: Solve the De Bruijn Graph from a String Problem.

**Sample Input**:

```
4
AAGATTCTCTAAGA
```

**Sample Output**:

```
AAG: AGA AGA
AGA: GAT
ATT: TTC
CTA: TAA
CTC: TCT
GAT: ATT
TAA: AAG
TCT: CTA CTC
TTC: TCT
```

````python
def PathGraph(k, text):
    path = []
    for i in range(len(text) - k + 1):
        kmer = text[i:i+k]
        prefix = kmer[:-1]
        suffix = kmer[1:]
        path.append((prefix, suffix))
    return path

def DeBruijn(k, text):
    graph = {}
    # We can use PathGraph to get the edges, then glue them
    edges = PathGraph(k, text)
    
    for prefix, suffix in edges:
        if prefix not in graph:
            graph[prefix] = []
        graph[prefix].append(suffix)
    return graph
````

````python
# Sample Input
k = 4
text = "AAGATTCTCTAAGA"

# Run the function
adjacency_list = DeBruijn(k, text)

# Print the output
for node in sorted(adjacency_list):
    neighbors = adjacency_list[node]
    # The sample output seems to have neighbors sorted
    print(f"{node}: {' '.join(sorted(neighbors))}")

# Test Assertion
expected_output = {
    'AAG': ['AGA', 'AGA'],
    'AGA': ['GAT'],
    'ATT': ['TTC'],
    'CTA': ['TAA'],
    'CTC': ['TCT'],
    'GAT': ['ATT'],
    'TAA': ['AAG'],
    'TCT': ['CTA', 'CTC'],
    'TTC': ['TCT']
}

# Normalize and check
formatted_output = {}
for k_node, v_neighbors in adjacency_list.items():
    formatted_output[k_node] = sorted(v_neighbors)

expected_normalized = {k: sorted(v) for k, v in expected_output.items()}

for k, v in expected_normalized.items():
    assert k in formatted_output, f"Key {k} missing"
    assert formatted_output[k] == v, f"For {k}, expected {v}, got {formatted_output[k]}"

print("Test passed!")
````

# De Bruijn Graph from k-mers Problem
Construct the de Bruijn graph from a set of k-mers.

**Input**: A collection of *k*-mers *Patterns*.

**Output**: The adjacency list of the de Bruijn graph DeBruijn(*Patterns*).

**Code Challenge**: Solve the De Bruijn Graph from k-mers Problem.

**Sample Input**:

```
GAGG CAGG GGGG GGGA CAGG AGGG GGAG
```

**Sample Output**:

```
AGG: GGG
CAG: AGG AGG
GAG: AGG
GGA: GAG
GGG: GGA GGG
```

````python
def DeBruijnFromKmers(patterns):
    graph = {}
    for kmer in patterns:
        prefix = kmer[:-1]
        suffix = kmer[1:]
        if prefix not in graph:
            graph[prefix] = []
        graph[prefix].append(suffix)
    return graph
````

````python
# Sample Input
patterns = ["GAGG", "CAGG", "GGGG", "GGGA", "CAGG", "AGGG", "GGAG"]

# Run the function
adjacency_list = DeBruijnFromKmers(patterns)

# Print the output
for node in sorted(adjacency_list):
    neighbors = adjacency_list[node]
    print(f"{node}: {' '.join(sorted(neighbors))}")

# Test Assertion
expected_output = {
    'AGG': ['GGG'],
    'CAG': ['AGG', 'AGG'],
    'GAG': ['AGG'],
    'GGA': ['GAG'],
    'GGG': ['GGA', 'GGG']
}

# Normalize and check
formatted_output = {}
for k_node, v_neighbors in adjacency_list.items():
    formatted_output[k_node] = sorted(v_neighbors)

expected_normalized = {k: sorted(v) for k, v in expected_output.items()}

for k, v in expected_normalized.items():
    assert k in formatted_output, f"Key {k} missing"
    assert formatted_output[k] == v, f"For {k}, expected {v}, got {formatted_output[k]}"

print("Test passed!")
````

````python
# Test Dataset
test_dataset_filename = 'dataset_30184_8.txt'
try:
    lines = get_dataset_lines(test_dataset_filename)
    patterns = []
    for line in lines:
        parts = line.strip().split()
        if parts:
            patterns.extend(parts)
    
    adjacency_list = DeBruijnFromKmers(patterns)
    
    # Print in the correct format
    for node in sorted(adjacency_list):
        neighbors = adjacency_list[node]
        print(f"{node}: {' '.join(sorted(neighbors))}")
            
except FileNotFoundError:
    print(f"File {test_dataset_filename} not found. Please download the dataset and check the filename.")
except Exception as e:
    print(f"An error occurred: {e}")
````
