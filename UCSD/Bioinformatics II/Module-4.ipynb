# Cyclopeptide Scoring Problem
Compute the score of a cyclic peptide against a spectrum.

**Input**: An amino acid string $Peptide$ and a collection of integers $Spectrum$.

**Output**: The score of $Peptide$ against $Spectrum$, $Score(Peptide, Spectrum)$.

**Code Challenge**: Solve the Cyclopeptide Scoring Problem.

**Sample Input**:

```
NQEL
0 99 113 114 128 227 257 299 355 356 370 371 484
```

**Sample Output**:

```
11
```

````python
def get_amino_acid_mass(filename='integer_mass_table.txt'):
    mass_table = {}
    with open(filename, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) == 2:
                mass_table[parts[0]] = int(parts[1])
    return mass_table

def CyclicSpectrum(Peptide, AminoAcidMass):
    PrefixMass = [0] * (len(Peptide) + 1)
    for i in range(1, len(Peptide) + 1):
        s = Peptide[i-1]
        PrefixMass[i] = PrefixMass[i-1] + AminoAcidMass[s]
    
    peptideMass = PrefixMass[len(Peptide)]
    CyclicSpectrum = [0]
    
    for i in range(len(Peptide)):
        for j in range(i + 1, len(Peptide) + 1):
            mass = PrefixMass[j] - PrefixMass[i]
            CyclicSpectrum.append(mass)
            if i > 0 and j < len(Peptide):
                CyclicSpectrum.append(peptideMass - mass)
                
    return sorted(CyclicSpectrum)
````

````python
from collections import Counter

def CyclopeptideScore(Peptide, Spectrum, AminoAcidMass):
    theoretical_spectrum = CyclicSpectrum(Peptide, AminoAcidMass)
    
    theo_counts = Counter(theoretical_spectrum)
    exp_counts = Counter(Spectrum)
    
    score = 0
    for mass in theo_counts:
        if mass in exp_counts:
            score += min(theo_counts[mass], exp_counts[mass])
            
    return score
````

````python
# Sample Input
peptide = "NQEL"
spectrum_str = "0 99 113 114 128 227 257 299 355 356 370 371 484"
spectrum = [int(x) for x in spectrum_str.split()]
amino_acid_mass = get_amino_acid_mass()

score = CyclopeptideScore(peptide, spectrum, amino_acid_mass)
print(score)

expected_output = 11
assert score == expected_output
print("Test passed!")
````

# Linear Score Problem
Compute the score of a linear peptide with respect to a spectrum.

**Input**: An amino acid string $Peptide$ and a collection of integers $Spectrum$.

**Output**: The linear score of $Peptide$ with respect to $Spectrum$, $LinearScore(Peptide, Spectrum)$.

**Code Challenge**: Solve the Linear Score Problem.

**Sample Input**:

```
NQEL
0 99 113 114 128 227 257 299 355 356 370 371 484
```

**Sample Output**:

```
8
```

````python
def LinearSpectrum(Peptide, AminoAcidMass):
    PrefixMass = [0] * (len(Peptide) + 1)
    for i in range(1, len(Peptide) + 1):
        s = Peptide[i-1]
        PrefixMass[i] = PrefixMass[i-1] + AminoAcidMass[s]
    
    LinearSpectrum = [0]
    for i in range(len(Peptide)):
        for j in range(i + 1, len(Peptide) + 1):
            LinearSpectrum.append(PrefixMass[j] - PrefixMass[i])
            
    return sorted(LinearSpectrum)

def LinearScore(Peptide, Spectrum, AminoAcidMass):
    theoretical_spectrum = LinearSpectrum(Peptide, AminoAcidMass)
    
    theo_counts = Counter(theoretical_spectrum)
    exp_counts = Counter(Spectrum)
    
    score = 0
    for mass in theo_counts:
        if mass in exp_counts:
            score += min(theo_counts[mass], exp_counts[mass])
            
    return score
````

# Trim Problem
Implement $Trim$ (reproduced below).

**Input**: A collection of peptides $Leaderboard$, a collection of integers $Spectrum$, and an integer $N$.

**Output**: The $N$ highest-scoring linear peptides on $Leaderboard$ with respect to $Spectrum$.

**Code Challenge**: Solve the Trim Problem.

**Sample Input**:

```
LAST ALST TLLT TQAS
0 71 87 101 113 158 184 188 259 271 372
2
```

**Sample Output**:

```
LAST ALST
```

````python
def Trim(Leaderboard, Spectrum, N, AminoAcidMass):
    scores = []
    for peptide in Leaderboard:
        score = LinearScore(peptide, Spectrum, AminoAcidMass)
        scores.append(score)
    
    # Sort Leaderboard based on scores in descending order
    # We zip them to keep pairs, sort, and unzip
    scored_leaderboard = sorted(zip(Leaderboard, scores), key=lambda x: x[1], reverse=True)
    
    Leaderboard = [x[0] for x in scored_leaderboard]
    LinearScores = [x[1] for x in scored_leaderboard]
    
    if len(Leaderboard) <= N:
        return Leaderboard
        
    # Handle ties
    cutoff_score = LinearScores[N-1] 
    
    trimmed_leaderboard = []
    for i in range(len(Leaderboard)):
        if LinearScores[i] >= cutoff_score:
            trimmed_leaderboard.append(Leaderboard[i])
        else:
            break 
            
    return trimmed_leaderboard
````

````python
# Sample Input for Trim
leaderboard = ["LAST", "ALST", "TLLT", "TQAS"]
spectrum_str = "0 71 87 101 113 158 184 188 259 271 372"
spectrum = [int(x) for x in spectrum_str.split()]
n = 2
amino_acid_mass = get_amino_acid_mass()

trimmed_leaderboard = Trim(leaderboard, spectrum, n, amino_acid_mass)
print(" ".join(trimmed_leaderboard))

expected_output = ["LAST", "ALST"]
assert trimmed_leaderboard == expected_output
print("Test passed!")
````

# Leaderboard Cyclopeptide Sequencing Problem
Implement $LeaderboardCyclopeptideSequencing$.

**Input**: An integer $N$ and a collection of integers $Spectrum$.

**Output**: $LeaderPeptide$ after running $LeaderboardCyclopeptideSequencing(Spectrum, N)$.

**Note**: Multiple solutions may exist. You may return any one.

**Code Challenge**: Solve the Leaderboard Cyclopeptide Sequencing Problem.

**Sample Input**:

```
10
0 71 113 129 147 200 218 260 313 331 347 389 460
```

**Sample Output**:

```
113-147-71-129
```

````python
def Mass(Peptide, AminoAcidMass):
    return sum(AminoAcidMass[aa] for aa in Peptide)

def Expand(Leaderboard, AminoAcidMass):
    expanded = []
    amino_acids = AminoAcidMass.keys()
    for peptide in Leaderboard:
        for aa in amino_acids:
            expanded.append(peptide + aa)
    return expanded

def LeaderboardCyclopeptideSequencing(Spectrum, N, AminoAcidMass):
    Leaderboard = [""]
    LeaderPeptide = ""
    ParentMass = max(Spectrum)
    
    while Leaderboard:
        Leaderboard = Expand(Leaderboard, AminoAcidMass)
        next_Leaderboard = []
        
        for Peptide in Leaderboard:
            peptide_mass = Mass(Peptide, AminoAcidMass)
            
            if peptide_mass == ParentMass:
                if CyclopeptideScore(Peptide, Spectrum, AminoAcidMass) > CyclopeptideScore(LeaderPeptide, Spectrum, AminoAcidMass):
                    LeaderPeptide = Peptide
            elif peptide_mass > ParentMass:
                continue # Remove Peptide (don't add to next_Leaderboard)
            
            next_Leaderboard.append(Peptide)
            
        Leaderboard = Trim(next_Leaderboard, Spectrum, N, AminoAcidMass)
        
    return LeaderPeptide
````

````python
# Sample Input for LeaderboardCyclopeptideSequencing
n = 10
spectrum_str = "0 71 113 129 147 200 218 260 313 331 347 389 460"
spectrum = [int(x) for x in spectrum_str.split()]
amino_acid_mass = get_amino_acid_mass()

leader_peptide = LeaderboardCyclopeptideSequencing(spectrum, n, amino_acid_mass)

# Convert to mass string for output
def peptide_to_mass_string(peptide, amino_acid_mass):
    masses = [str(amino_acid_mass[aa]) for aa in peptide]
    return "-".join(masses)

output = peptide_to_mass_string(leader_peptide, amino_acid_mass)
print(f"Actual: {output}")

# Verify against sample output (order might differ, but masses should be consistent)
expected_output = "113-147-71-129"
# Note: There might be permutations or other valid solutions.
# The sample output is just one valid solution.
print(f"Expected: {expected_output}")

# Check if the score matches the expected solution's score
expected_peptide_masses = [int(x) for x in expected_output.split('-')]
# We can't easily reconstruct the string from masses without ambiguity, 
# but we can check if our result has the same score.
print(f"Score of result: {CyclopeptideScore(leader_peptide, spectrum, amino_acid_mass)}")
````

# Spectral Convolution Problem
Compute the convolution of a spectrum.

**Input**: A collection of integers $Spectrum$ in increasing order.

**Output**: The list of elements in the convolution of $Spectrum$. If an element has multiplicity $k$, it should appear exactly $k$ times; you may return the elements in any order.

**Code Challenge**: Solve the Spectral Convolution Problem.

**Hint**: The convolution that you return should not contain any zeroes.

**Sample Input**:

```
0 137 186 323
```

**Sample Output**:

```
137 137 186 186 323 49
```

````python
def SpectralConvolution(Spectrum):
    convolution = []
    Spectrum.sort()
    for i in range(len(Spectrum)):
        for j in range(i):
            diff = Spectrum[i] - Spectrum[j]
            if diff > 0:
                convolution.append(diff)
    return convolution
````

````python
# Sample Input
spectrum_str = "0 137 186 323"
spectrum = [int(x) for x in spectrum_str.split()]

convolution = SpectralConvolution(spectrum)
print(" ".join(map(str, convolution)))

# Verify against sample output (order may vary)
expected_output = [137, 137, 186, 186, 323, 49]
assert sorted(convolution) == sorted(expected_output)
print("Test passed!")
````

# Convolution Cyclopeptide Sequencing Problem
Implement $ConvolutionCyclopeptideSequencing$.

**Input**: An integer $M$, an integer $N$, and a collection of (possibly repeated) integers $Spectrum$.

**Output**: A cyclic peptide $LeaderPeptide$ with amino acids taken only from the top $M$ elements (and ties) of the convolution of $Spectrum$ that fall between 57 and 200, and where the size of $Leaderboard$ is restricted to the top $N$ (and ties).

**Code Challenge**: Solve the Convolution Cyclopeptide Sequencing Problem.

**Sample Input**:

```
20
60
57 57 71 99 129 137 170 186 194 208 228 265 285 299 307 323 356 364 394 422 493
```

**Sample Output**:

```
99-71-137-57-72-57
```

````python
def ConvolutionCyclopeptideSequencing(Spectrum, M, N):
    # 1. Compute Spectral Convolution
    convolution = SpectralConvolution(Spectrum)
    
    # 2. Filter masses between 57 and 200
    filtered_convolution = [m for m in convolution if 57 <= m <= 200]
    
    # 3. Find the top M elements (with ties)
    mass_counts = Counter(filtered_convolution)
    sorted_masses = sorted(mass_counts.items(), key=lambda x: x[1], reverse=True)
    
    top_M_masses = []
    if len(sorted_masses) > 0:
        cutoff_count = sorted_masses[min(M, len(sorted_masses)) - 1][1]
        for mass, count in sorted_masses:
            if count >= cutoff_count:
                top_M_masses.append(mass)
            else:
                break
    
    # 4. Run LeaderboardCyclopeptideSequencing with the restricted alphabet
    # We need to use the tuple version since masses are integers
    leader_peptides = LeaderboardCyclopeptideSequencingAllTuple(Spectrum, N, top_M_masses)
    
    # Return one of the leader peptides (as per problem statement)
    if leader_peptides:
        return list(leader_peptides)[0]
    return ()
````

````python
# Sample Input
m = 20
n = 60
spectrum_str = "57 57 71 99 129 137 170 186 194 208 228 265 285 299 307 323 356 364 394 422 493"
spectrum = [int(x) for x in spectrum_str.split()]

leader_peptide = ConvolutionCyclopeptideSequencing(spectrum, m, n)

# Format output
output = "-".join(map(str, leader_peptide))
print(output)

# Verify score against expected output
expected_output_str = "99-71-137-57-72-57"
expected_peptide = tuple(map(int, expected_output_str.split('-')))

score_result = CyclopeptideScoreTuple(leader_peptide, spectrum)
score_expected = CyclopeptideScoreTuple(expected_peptide, spectrum)

print(f"Score of result: {score_result}")
print(f"Score of expected: {score_expected}")

# Note: The problem asks for *a* cyclic peptide. There might be multiple with the same max score.
# We check if our score is at least as good as the sample output.
assert score_result >= score_expected
print("Test passed!")
````

````python
# Final Challenge: Real Spectrum
filename = 'real_spectrum.txt'
try:
    with open(filename, 'r') as f:
        raw_spectrum = [float(x) for x in f.read().strip().split()]

    # Adjust for charge +1 (M+H) -> Mass = m/z - 1 (approx)
    # We subtract 1.0 and then round to nearest integer
    spectrum = [int(round(x - 1.0)) for x in raw_spectrum]
    
    # Also, we should probably add 0 to the spectrum if it's not there, 
    # as our algorithms assume 0 is present (prefix mass of empty peptide).
    if 0 not in spectrum:
        spectrum.append(0)
    spectrum.sort()

    # Parameters
    N = 1000
    M = 20 
    
    # Run ConvolutionCyclopeptideSequencing
    # Note: We need to use the version that returns all top candidates or just one.
    # The problem asks for "the peptide", implying a single best one or any valid one.
    
    # 1. Compute Spectral Convolution on the integer spectrum
    convolution = SpectralConvolution(spectrum)
    
    # 2. Filter masses between 57 and 200
    filtered_convolution = [m for m in convolution if 57 <= m <= 200]
    
    # 3. Find top M masses
    mass_counts = Counter(filtered_convolution)
    sorted_masses = sorted(mass_counts.items(), key=lambda x: x[1], reverse=True)
    
    print("Top 20 masses from convolution:")
    for m, c in sorted_masses[:20]:
        print(f"Mass: {m}, Count: {c}")
        
    top_M_masses = []
    if len(sorted_masses) > 0:
        cutoff_count = sorted_masses[min(M, len(sorted_masses)) - 1][1]
        for mass, count in sorted_masses:
            if count >= cutoff_count:
                top_M_masses.append(mass)
            else:
                break
    
    print(f"Selected {len(top_M_masses)} masses for alphabet: {top_M_masses}")

    # 4. Run LeaderboardCyclopeptideSequencing
    leader_peptides = LeaderboardCyclopeptideSequencingAllTuple(spectrum, N, top_M_masses)
    
    # Format output
    formatted_peptides = []
    for peptide in leader_peptides:
        masses_str = [str(m) for m in peptide]
        formatted_peptides.append("-".join(masses_str))
    
    formatted_peptides.sort()
    
    print("\nFound Peptides:")
    print(" ".join(formatted_peptides))
    
    if leader_peptides:
        best_peptide = list(leader_peptides)[0]
        print(f"Max Score: {CyclopeptideScoreTuple(best_peptide, spectrum)}")
        
        # Output as space-separated integers as requested
        print("\nFinal Answer:")
        print(" ".join(map(str, best_peptide)))

except FileNotFoundError:
    print(f"File {filename} not found.")
````
