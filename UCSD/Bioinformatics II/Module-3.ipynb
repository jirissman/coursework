The following problem asks you to find the translation of an RNA string into an amino acid string.

Protein Translation Problem: Translate an RNA string into an amino acid string.

Input: An RNA string Pattern and the array GeneticCode.
Output: The translation of Pattern into an amino acid string Peptide.
Code Challenge: Solve the Protein Translation Problem.

Notes:

The "Stop" codon should not be translated, as shown in the sample below.
For your convenience, we provide a downloadable RNA codon table indicating which codons encode which amino acids.


Sample Input:

AUGGCCAUGGCGCCCAGAACUGAGAUCAAUAGUACCCGUAUUAACGGGUGA
Sample Output:

MAMAPRTEINSTRING


```python
def get_rna_codon_table(filename='RNA_codon_table_1.txt'):
    codon_table = {}
    with open(filename, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) == 0: continue
            if len(parts) == 2:
                codon_table[parts[0]] = parts[1]
            else:
                codon_table[parts[0]] = None # Stop codon
    return codon_table

def ProteinTranslation(rna_pattern, genetic_code):
    peptide = ""
    for i in range(0, len(rna_pattern), 3):
        codon = rna_pattern[i:i+3]
        if len(codon) < 3:
            break
        amino_acid = genetic_code.get(codon)
        if amino_acid is None:
            return peptide # Stop translation
        peptide += amino_acid
    return peptide
```

```python
# Sample Input
sample_input = "AUGGCCAUGGCGCCCAGAACUGAGAUCAAUAGUACCCGUAUUAACGGGUGA"
genetic_code = get_rna_codon_table()

peptide = ProteinTranslation(sample_input, genetic_code)
print(peptide)

expected_output = "MAMAPRTEINSTRING"
assert peptide == expected_output
print("Test passed!")
```

Peptide Encoding Problem: Find substrings of a genome encoding a given amino acid sequence.

Input: A DNA string Text, an amino acid string Peptide, and the array GeneticCode.
Output: All substrings of Text encoding Peptide (if any such substrings exist).
Code Challenge: Solve the Peptide Encoding Problem. Click here for the RNA codon table corresponding to the array GeneticCode.

Note: The solution may contain repeated strings if the same string occurs more than once as a substring of Text and encodes Peptide.

Sample Input:

ATGGCCATGGCCCCCAGAACTGAGATCAATAGTACCCGTATTAACGGGTGA
MA
Sample Output:

ATGGCC
GGCCAT
ATGGCC

```python
def ReverseComplement(pattern):
    complement = {'A': 'T', 'C': 'G', 'G': 'C', 'T': 'A'}
    return "".join(complement[base] for base in reversed(pattern))

def DNATranscription(dna):
    return dna.replace('T', 'U')

def PeptideEncoding(text, peptide, genetic_code):
    k = len(peptide) * 3
    results = []
    for i in range(len(text) - k + 1):
        pattern = text[i:i+k]
        
        # Check forward strand
        rna = DNATranscription(pattern)
        if ProteinTranslation(rna, genetic_code) == peptide:
            results.append(pattern)
            continue 
            
        # Check reverse complement strand
        rc_pattern = ReverseComplement(pattern)
        rna_rc = DNATranscription(rc_pattern)
        if ProteinTranslation(rna_rc, genetic_code) == peptide:
            results.append(pattern)
            
    return results
```

```python
# Sample Input
text = "ATGGCCATGGCCCCCAGAACTGAGATCAATAGTACCCGTATTAACGGGTGA"
peptide = "MA"
genetic_code = get_rna_codon_table()

results = PeptideEncoding(text, peptide, genetic_code)
for res in results:
    print(res)

expected_output = ["ATGGCC", "GGCCAT", "ATGGCC"]
assert results == expected_output
print("Test passed!")
```

Code Challenge: Implement LinearSpectrum.

Input: An amino acid string Peptide.
Output: The linear spectrum of Peptide.

Sample Input:

NQEL
Sample Output:

0 113 114 128 129 242 242 257 370 371 484

```python
def get_amino_acid_mass(filename='integer_mass_table.txt'):
    mass_table = {}
    with open(filename, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) == 2:
                mass_table[parts[0]] = int(parts[1])
    return mass_table

def LinearSpectrum(Peptide, AminoAcidMass):
    PrefixMass = [0] * (len(Peptide) + 1)
    for i in range(1, len(Peptide) + 1):
        s = Peptide[i-1]
        PrefixMass[i] = PrefixMass[i-1] + AminoAcidMass[s]
    
    LinearSpectrum = [0]
    for i in range(len(Peptide)):
        for j in range(i + 1, len(Peptide) + 1):
            LinearSpectrum.append(PrefixMass[j] - PrefixMass[i])
            
    return sorted(LinearSpectrum)
```

```python
# Sample Input
peptide = "NQEL"
amino_acid_mass = get_amino_acid_mass()
spectrum = LinearSpectrum(peptide, amino_acid_mass)
print(*spectrum)

expected_output = [0, 113, 114, 128, 129, 242, 242, 257, 370, 371, 484]
assert spectrum == expected_output
print("Test passed!")
```

Generating Theoretical Spectrum Problem: Generate the theoretical spectrum of a cyclic peptide.

Input: An amino acid string Peptide.
Output: Cyclospectrum(Peptide).
Code Challenge: Solve the Generating Theoretical Spectrum Problem.

Sample Input:

LEQN
Sample Output:

0 113 114 128 129 227 242 242 257 355 356 370 371 484

```python
def CyclicSpectrum(Peptide, AminoAcidMass):
    PrefixMass = [0] * (len(Peptide) + 1)
    for i in range(1, len(Peptide) + 1):
        s = Peptide[i-1]
        PrefixMass[i] = PrefixMass[i-1] + AminoAcidMass[s]
    
    peptideMass = PrefixMass[len(Peptide)]
    CyclicSpectrum = [0]
    
    for i in range(len(Peptide)):
        for j in range(i + 1, len(Peptide) + 1):
            mass = PrefixMass[j] - PrefixMass[i]
            CyclicSpectrum.append(mass)
            if i > 0 and j < len(Peptide):
                CyclicSpectrum.append(peptideMass - mass)
                
    return sorted(CyclicSpectrum)
```

```python
# Sample Input
peptide = "LEQN"
amino_acid_mass = get_amino_acid_mass()
spectrum = CyclicSpectrum(peptide, amino_acid_mass)
print(*spectrum)

expected_output = [0, 113, 114, 128, 129, 227, 242, 242, 257, 355, 356, 370, 371, 484]
assert spectrum == expected_output
print("Test passed!")
```

Counting Peptides with Given Mass Problem: Compute the number of peptides of given mass.

Input: An integer m.
Output: The number of linear peptides having integer mass m.
Exercise Break (Optional): Solve the Counting Peptides with Given Mass Problem. The integer mass table is reproduced below; recall that we assume that peptides are formed from only 18 amino acid masses. That is, I/L are considered the same and K/Q are considered the same, so we would count AIKD and ALQD as just one peptide, not two.

Sample Input:

1024
Sample Output:

14712706211

```python
def CountPeptidesWithGivenMass(m, mass_table):
    unique_masses = set(mass_table.values())
    dp = {0: 1}
    
    for i in range(1, m + 1):
        count = 0
        for mass in unique_masses:
            if i >= mass:
                count += dp.get(i - mass, 0)
        dp[i] = count
        
    return dp[m]
```

```python
# Sample Input
m = 1024
amino_acid_mass = get_amino_acid_mass()
count = CountPeptidesWithGivenMass(m, amino_acid_mass)
print(count)

expected_output = 14712706211
assert count == expected_output
print("Test passed!")
```

# Exercise Break: Find C
For large $m$, the number of peptides with given integer mass $m$ can be approximated as $k \cdot C^m$, where $k$ and $C$ are constants.
We can estimate $C$ by calculating the ratio $N(m) / N(m-1)$ for a large $m$.


```python
m_large = 4000
amino_acid_mass = get_amino_acid_mass()
unique_masses = set(amino_acid_mass.values())
dp = {0: 1}

for i in range(1, m_large + 1):
    count = 0
    for mass in unique_masses:
        if i >= mass:
            count += dp.get(i - mass, 0)
    dp[i] = count

C = dp[m_large] / dp[m_large-1]
print(f"Estimated C for m={m_large}: {C}")
```

# Exercise Break: Count Subpeptides
How many subpeptides does a linear peptide of given length $n$ have? (Include the empty peptide and the entire peptide.)

Input: An integer $n$.
Output: The number of subpeptides of a linear peptide of length $n$.

Sample Input:
4

Sample Output:
11

```python
def CountSubpeptidesLinear(n):
    return n * (n + 1) // 2 + 1
```

Code Challenge: Implement CyclopeptideSequencing (pseudocode reproduced below).

Sample Input:

0 113 128 186 241 299 314 427
Sample Output:

186-128-113 186-113-128 128-186-113 128-113-186 113-186-128 113-128-186

```python
from collections import Counter

def LinearSpectrumMasses(Peptide):
    PrefixMass = [0] * (len(Peptide) + 1)
    for i in range(1, len(Peptide) + 1):
        PrefixMass[i] = PrefixMass[i-1] + Peptide[i-1]
    
    LinearSpectrum = [0]
    for i in range(len(Peptide)):
        for j in range(i + 1, len(Peptide) + 1):
            LinearSpectrum.append(PrefixMass[j] - PrefixMass[i])
            
    return sorted(LinearSpectrum)

def CyclicSpectrumMasses(Peptide):
    PrefixMass = [0] * (len(Peptide) + 1)
    for i in range(1, len(Peptide) + 1):
        PrefixMass[i] = PrefixMass[i-1] + Peptide[i-1]
    
    peptideMass = PrefixMass[len(Peptide)]
    CyclicSpectrum = [0]
    
    for i in range(len(Peptide)):
        for j in range(i + 1, len(Peptide) + 1):
            mass = PrefixMass[j] - PrefixMass[i]
            CyclicSpectrum.append(mass)
            if i > 0 and j < len(Peptide):
                CyclicSpectrum.append(peptideMass - mass)
                
    return sorted(CyclicSpectrum)

def Consistent(Peptide, Spectrum):
    linear_spec = LinearSpectrumMasses(Peptide)
    spec_counts = Counter(Spectrum)
    linear_counts = Counter(linear_spec)
    for mass, count in linear_counts.items():
        if spec_counts[mass] < count:
            return False
    return True

def Expand(CandidatePeptides, unique_masses):
    new_candidates = set()
    for peptide in CandidatePeptides:
        for mass in unique_masses:
            new_candidates.add(peptide + (mass,))
    return new_candidates

def CyclopeptideSequencing(Spectrum):
    amino_acid_mass = get_amino_acid_mass()
    unique_masses = sorted(list(set(amino_acid_mass.values())))
    
    CandidatePeptides = {()}
    FinalPeptides = []
    
    parent_mass = max(Spectrum)
    Spectrum.sort() 
    
    while CandidatePeptides:
        CandidatePeptides = Expand(CandidatePeptides, unique_masses)
        next_CandidatePeptides = set()
        
        for Peptide in CandidatePeptides:
            peptide_mass = sum(Peptide)
            if peptide_mass == parent_mass:
                if CyclicSpectrumMasses(Peptide) == Spectrum:
                    FinalPeptides.append(Peptide)
            elif Consistent(Peptide, Spectrum):
                next_CandidatePeptides.add(Peptide)
        
        CandidatePeptides = next_CandidatePeptides
        
    return FinalPeptides
```

```python
# Sample Input
sample_input = "0 113 128 186 241 299 314 427"
spectrum = [int(x) for x in sample_input.split()]

final_peptides = CyclopeptideSequencing(spectrum)

# Format output
formatted_output = []
for peptide in final_peptides:
    formatted_output.append("-".join(map(str, peptide)))

# Sort to ensure consistent order for comparison
formatted_output.sort()
print(" ".join(formatted_output))

expected_output = "186-128-113 186-113-128 128-186-113 128-113-186 113-186-128 113-128-186"
expected_output_list = sorted(expected_output.split())

assert formatted_output == expected_output_list
print("Test passed!")
```
