## RMSD Tutorial Example

This is a simple example of calculating the RMSD between a predicted SARS-2 spike protein model using SWISS (swiss1) and an experimentally determined model from the Protein Data Bank (6vxx). This notebook directly follows the RMSD tutorial from Module 3 of *Multiscale Biological Modeling: A Free Online Course*.

**Make sure that this notebook is in the same file directory as the structures files.**

```python
from prody import parsePDB, calcRMSD, matchChains, calcTransformation
```

```python
def printMatch(match):
    print('Chain 1 : {}'.format(match[0]))
    print('Chain 2 : {}'.format(match[1]))
    print('Length : {}'.format(len(match[0])))
    print('Seq identity: {}'.format(match[2]))
    print('Seq overlap : {}'.format(match[3]))
    if len(match[0]) > 0:
        print('RMSD : {}\n'.format(calcRMSD(match[0], match[1])))
```

```python
filepath = '/Users/jirissman/Documents/coursework/UCSD/Capstone/SWISS-MODEL/model_02.pdb'
struct1 = parsePDB('6crx.pdb')
# struct1 = parsePDB('swiss1.pdb')
struct1 = parsePDB(filepath)
```

```python
struct2 = parsePDB('6vxx.pdb')
```

```python
matches = matchChains(struct1.select('protein'), struct2.select('protein'), seqid=75, overlap=80) # type: ignore
```

```python
for match in matches:
    printMatch(match)
```

### Calculating RMSD of two chains

```python
first_ca = matches[0][0]
second_ca = matches[0][1]
calcTransformation(first_ca, second_ca).apply(first_ca);
calcRMSD(first_ca,second_ca)
```

### Merging multiple chains to compute RMSD of an overall structure

```python
first_ca = matches[0][0] + matches[4][0] + matches[8][0]
second_ca = matches [0][1] + matches[4][1] + matches[8][1]
calcTransformation(first_ca, second_ca).apply(first_ca);
calcRMSD(first_ca, second_ca)
```
