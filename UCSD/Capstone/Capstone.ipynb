```python
import numpy as np

def calculate_rmsd_centered(coords1, coords2):
    """
    Calculates the RMSD between two structures after aligning their
    centers of mass (translation only).
    """
    # Convert inputs to NumPy arrays for vectorized math
    P = np.array(coords1)
    Q = np.array(coords2)

    # Ensure both structures have the same number of points
    if P.shape != Q.shape:
        raise ValueError("Structures must have the same number of points.")

    # 1. Calculate the Centers of Mass (Centroids)
    center_P = np.mean(P, axis=0)
    center_Q = np.mean(Q, axis=0)

    # 2. Align structures by shifting them to the origin
    # (Subtract the centroid from each point)
    P_centered = P - center_P
    Q_centered = Q - center_Q

    # 3. Calculate the squared Euclidean distances between corresponding points
    # (x1-x2)^2 + (y1-y2)^2
    diff = P_centered - Q_centered
    squared_distances = np.sum(diff**2, axis=1)

    # 4. Calculate the Mean Squared Distance
    mean_sq_dist = np.mean(squared_distances)

    # 5. Take the square root to get RMSD
    rmsd = np.sqrt(mean_sq_dist)

    return rmsd, center_P, center_Q, P_centered, Q_centered, squared_distances, mean_sq_dist

# --- Input Data ---

structure_1 = [
    [1, 2], [2, 2], [3, 5], [5, 3], 
    [2, 1], [7, 2], [5, 4], [7, 5]
]

structure_2 = [
    [1, 3], [2, 2], [4, 5], [5, 3], 
    [4, 1], [7, 2], [6, 3], [7, 4]
]

# --- Calculation ---

rmsd_value, center_P, center_Q, P_centered, Q_centered, squared_distances, mean_sq_dist = calculate_rmsd_centered(structure_1, structure_2)

print(f"RMSD: {rmsd_value}")
print(f"Center of Structure 1: {center_P}")
print(f"Center of Structure 2: {center_Q}")
print(f"Squared Distances: {squared_distances}")
print(f"Mean Squared Distance: {mean_sq_dist}")
print(f"Centered Structure 1:\n{P_centered}")
print(f"Centered Structure 2:\n{Q_centered}")
```
