# Probability of a Hidden Path Problem

Compute the probability of a hidden path.

**Input**: A hidden path $\pi$ in an HMM $(\Sigma, States, Transition, Emission)$.

**Output**: The probability of this path, $Pr(\pi)$.

**Code Challenge**: Solve the Probability of a Hidden Path Problem.

**Input**: A hidden path $\pi$ followed by the states $States$ and transition matrix $Transition$ of an HMM $(\Sigma, States, Transition, Emission)$.

**Output**: The probability of this path, $Pr(\pi)$.

**Note**: You may assume that transitions from the initial state occur with equal probability.

**Sample Input**:

```
ABABBBAAAA
--------
A B
--------
	A	B
A	0.377	0.623
B	0.26	0.74
```

**Sample Output**:

```
0.000384928691755
```

````python
def parse_hidden_path_problem(lines):
    lines = [l.strip() for l in lines]
    path = lines[0]
    # Separator
    states = lines[2].split()
    # Separators and Header
    
    transitions = {}
    matrix_lines = lines[5:]
    
    for line in matrix_lines:
        if not line: continue
        parts = line.split()
        row_state = parts[0]
        probs = [float(x) for x in parts[1:]]
        transitions[row_state] = {}
        for i, col_state in enumerate(states):
            transitions[row_state][col_state] = probs[i]
            
    return path, states, transitions

def probability_of_hidden_path(path, states, transitions):
    # Initialize uniform probability
    prob = 1.0 / len(states)
    
    for i in range(len(path) - 1):
        prev_node = path[i]
        curr_node = path[i+1]
        prob *= transitions[prev_node][curr_node]
        
    return prob
````

````python
### Sample Input Execution
sample_input = """
ABABBBAAAA
--------
A B
--------
	A	B
A	0.377	0.623
B	0.26	0.74
""".strip().split('\n')

path, states, transitions = parse_hidden_path_problem(sample_input)
result = probability_of_hidden_path(path, states, transitions)
print(result)

### Verification
expected_output = 0.000384928691755
assert abs(result - expected_output) < 1e-10, f"Expected {expected_output}, got {result}"
print("Sample test passed!")
````

# Probability of an Outcome Given a Hidden Path Problem

Compute the probability that an HMM will emit a string given its hidden path.

**Input**: A string $x = x_1 \dots x_n$ emitted by an HMM $(\Sigma, States, Transition, Emission)$ and a hidden path $\pi = \pi_1 \dots \pi_n$.

**Output**: The conditional probability $Pr(x|\pi)$ that $x$ will be emitted given that the HMM follows the hidden path $\pi$.

**Code Challenge**: Solve the Probability of an Outcome Given a Hidden Path Problem.

**Input**: A string $x$, followed by the alphabet from which $x$ was constructed, followed by a hidden path $\pi$, followed by the states $States$ and emission matrix $Emission$ of an HMM $(\Sigma, States, Transition, Emission)$.

**Output**: The conditional probability $Pr(x|\pi)$ that $x$ will be emitted given that the HMM follows the hidden path $\pi$.

**Note**: You may assume that transitions from the initial state occur with equal probability.

**Sample Input**:

```
zzzyxyyzzx
--------
x y z
--------
BAAAAAAAAA
--------
A B
--------
	x	y	z
A	0.176	0.596	0.228
B	0.225	0.572	0.203
```

**Sample Output**:

```
3.59748954746e-06
```

````python
def parse_outcome_given_path(lines):
    lines = [l.strip() for l in lines]
    x = lines[0]
    # Separator
    alphabet = lines[2].split()
    # Separator
    path = lines[4]
    # Separator
    states = lines[6].split()
    # Separator and Header
    
    emissions = {}
    matrix_lines = lines[9:]
    
    for line in matrix_lines:
        if not line: continue
        parts = line.split()
        row_state = parts[0]
        probs = [float(val) for val in parts[1:]]
        emissions[row_state] = {}
        for i, symbol in enumerate(alphabet):
            emissions[row_state][symbol] = probs[i]
            
    return x, alphabet, path, states, emissions

def probability_outcome_given_path(x, path, emissions):
    prob = 1.0
    for i in range(len(x)):
        symbol = x[i]
        state = path[i]
        prob *= emissions[state][symbol]
    return prob
````

````python
### Sample Input Execution
sample_input = """
zzzyxyyzzx
--------
x y z
--------
BAAAAAAAAA
--------
A B
--------
	x	y	z
A	0.176	0.596	0.228
B	0.225	0.572	0.203
""".strip().split('\n')

x, alphabet, path, states, emissions = parse_outcome_given_path(sample_input)
result = probability_outcome_given_path(x, path, emissions)
print(result)

### Verification
expected_output = 3.59748954746e-06
assert abs(result - expected_output) < 1e-15, f"Expected {expected_output}, got {result}"
print("Sample test passed!")
````

# Decoding Problem

**Code Challenge**: Implement the Viterbi algorithm solving the Decoding Problem.

**Input**: A string $x$, followed by the alphabet from which $x$ was constructed, followed by the states $States$, transition matrix $Transition$, and emission matrix $Emission$ of an HMM $(\Sigma, States, Transition, Emission)$.

**Output**: A path that maximizes the (unconditional) probability $Pr(x, \pi)$ over all possible paths $\pi$.

**Note**: You may assume that transitions from the initial state occur with equal probability.

**Sample Input**:

```
xyxzzxyxyy
--------
x y z
--------
A B
--------
	A	B
A	0.641	0.359
B	0.729	0.271
--------
	x	y	z
A	0.117	0.691	0.192	
B	0.097	0.42	0.483
```

**Sample Output**:

```
AAABBAAAAA
```

````python
def parse_viterbi_input(text_lines):
    # Split input sections
    full_text = "\n".join([l.strip() for l in text_lines])
    sections = full_text.split('--------')
    
    x = sections[0].strip()
    alphabet = sections[1].strip().split()
    states = sections[2].strip().split()
    
    # Clean lines
    trans_lines = sections[3].strip().split('\n')
    trans_lines = [l for l in trans_lines if l.strip()]
    
    transitions = {}
    for line in trans_lines[1:]:
        parts = line.split()
        row_state = parts[0]
        probs = [float(v) for v in parts[1:]]
        transitions[row_state] = {}
        for i, col_state in enumerate(states):
            transitions[row_state][col_state] = probs[i]
            
    # Parse Emission Matrix
    emit_lines = sections[4].strip().split('\n')
    emit_lines = [l for l in emit_lines if l.strip()]
    
    emissions = {}
    for line in emit_lines[1:]:
        parts = line.split()
        row_state = parts[0]
        probs = [float(v) for v in parts[1:]]
        emissions[row_state] = {}
        for i, symbol in enumerate(alphabet):
            emissions[row_state][symbol] = probs[i]
            
    return x, alphabet, states, transitions, emissions

def viterbi_algorithm(x, alphabet, states, transitions, emissions):
    n = len(x)
    # Dynamic programming tables
    T1 = [{state: 0.0 for state in states} for _ in range(n)]
    T2 = [{state: None for state in states} for _ in range(n)]
    
    # Base Case: Assumes uniform start probabilities
    init_prob = 1.0 / len(states)
    for state in states:
        T1[0][state] = init_prob * emissions[state][x[0]]
        
    # Recursion
    for i in range(1, n):
        for j in states: # current state
            max_val = -1.0
            max_prev = None
            
            emission_prob = emissions[j][x[i]]
            
            for k in states: # previous state
                # Calculate probability
                prob = T1[i-1][k] * transitions[k][j] * emission_prob
                
                if prob > max_val:
                    max_val = prob
                    max_prev = k
            
            T1[i][j] = max_val
            T2[i][j] = max_prev
            
    # Backtrack
    last_step_max = -1.0
    last_state = states[0] # Default
    
    for state in states:
        if T1[n-1][state] >= last_step_max:
            last_step_max = T1[n-1][state]
            last_state = state
            
    # Reconstruct path
    path = [last_state]
    curr = last_state
    for i in range(n-1, 0, -1):
        prev = T2[i][curr]
        if prev is None: # Should not happen in connected HMM
            prev = states[0] 
        path.append(prev)
        curr = prev
        
    return "".join(path[::-1])
````

````python
### Sample Input Execution
sample_input = """
xyxzzxyxyy
--------
x y z
--------
A B
--------
	A	B
A	0.641	0.359
B	0.729	0.271
--------
	x	y	z
A	0.117	0.691	0.192	
B	0.097	0.42	0.483
""".strip().split('\n')

x_samp, alpha_samp, states_samp, trans_samp, emit_samp = parse_viterbi_input(sample_input)
result_path = viterbi_algorithm(x_samp, alpha_samp, states_samp, trans_samp, emit_samp)
print(result_path)

### Verification
expected_output = "AAABBAAAAA"
assert result_path == expected_output, f"Expected {expected_output}, got {result_path}"
print("Sample test passed!")
````

# Outcome Likelihood Problem

**Code Challenge**: Solve the Outcome Likelihood Problem.

**Input**: A string $x$, followed by the alphabet from which $x$ was constructed, followed by the states $States$, transition matrix $Transition$, and emission matrix $Emission$ of an HMM $(\Sigma, States, Transition, Emission)$.

**Output**: The probability $Pr(x)$ that the HMM emits $x$.

**Note**: You may assume that transitions from the initial state occur with equal probability.

Extra Dataset and Test Datasets

**Sample Input**:

```
xzyyzzyzyy
--------
x y z
--------
A B
--------
	A	B
A	0.303	0.697 
B	0.831	0.169 
--------
	x	y	z
A	0.533	0.065	0.402 
B	0.342	0.334	0.324
```

**Sample Output**:

```
1.1005510319694847e-06
```

````python
def outcome_likelihood(x, alphabet, states, transitions, emissions):
    n = len(x)
    # DP table for probabilities
    forward = [{state: 0.0 for state in states} for _ in range(n)]
    
    # Base case: Assumes uniform start probabilities
    init_prob = 1.0 / len(states)
    for state in states:
        forward[0][state] = init_prob * emissions[state][x[0]]
        
    # Recursion
    for i in range(1, n):
        for curr_state in states:
            sum_prev = 0.0
            emission_prob = emissions[curr_state][x[i]]
            
            for prev_state in states:
                # Sum previous probabilities
                sum_prev += forward[i-1][prev_state] * transitions[prev_state][curr_state]
            
            forward[i][curr_state] = sum_prev * emission_prob
            
    # Total probability
    total_prob = sum(forward[n-1].values())
    return total_prob
````

````python
### Sample Input Execution
sample_input = """
xzyyzzyzyy
--------
x y z
--------
A B
--------
	A	B
A	0.303	0.697 
B	0.831	0.169 
--------
	x	y	z
A	0.533	0.065	0.402 
B	0.342	0.334	0.324
""".strip().split('\n')

x_samp, alpha_samp, states_samp, trans_samp, emit_samp = parse_viterbi_input(sample_input)
result = outcome_likelihood(x_samp, alpha_samp, states_samp, trans_samp, emit_samp)
print(result)

### Verification
expected_output = 1.1005510319694847e-06
assert abs(result - expected_output) < 1e-15, f"Expected {expected_output}, got {result}"
print("Sample test passed!")
````
