# Trie Construction Problem

Construct a trie from a set of patterns.

**Input**: A collection of strings $Patterns$.

**Output**: $Trie(Patterns)$.

The most obvious way to construct $Trie(Patterns)$ is by iteratively adding each string from $Patterns$ to the growing trie, as implemented by the following algorithm.

```plaintext
TrieConstruction(Patterns)
    Trie ← a graph consisting of a single node root
    for each string Pattern in Patterns
        currentNode ← root
        for i ← 0 to |Pattern| - 1
            currentSymbol ← Pattern[i]
            if there is an outgoing edge from currentNode with label currentSymbol
                currentNode ← ending node of this edge
            else
                add a new node newNode to Trie
                add a new edge from currentNode to newNode with label currentSymbol
                currentNode ← newNode
    return Trie
```

**Code Challenge**: Solve the Trie Construction Problem.

**Input**: A space-separated collection of strings $Patterns$.

**Output**: The adjacency list corresponding to $Trie(Patterns)$, in the following format. If $Trie(Patterns)$ has $n$ nodes, first label the root with $0$ and then label the remaining nodes with the integers $1$ through $n - 1$ in any order you like. Each edge of the adjacency list of $Trie(Patterns)$ will be encoded by a triple: the first two members of the triple must be the integers labeling the initial and terminal nodes of the edge, respectively; the third member of the triple must be the symbol labeling the edge.

**Sample Input**:

```
ATAGA ATC GAT
```

**Sample Output**:

```
0 1 A
1 2 T
2 3 A
3 4 G
4 5 A
2 6 C
0 7 G
7 8 A
8 9 T
```

````python
def TrieConstruction(patterns):
    """
    Constructs a trie from a collection of patterns.
    
    Args:
        patterns: A list of strings.
        
    Returns:
        A tuple containing:
        1. A list of tuples (u, v, symbol) representing the adjacency list.
        2. A dictionary representing the graph structure {node: {symbol: next_node}}.
    """
    trie = []
    # Graph structure to keep track of existing paths: node -> {symbol -> next_node}
    graph = {0: {}}
    next_node = 1
    
    for pattern in patterns:
        current_node = 0
        for symbol in pattern:
            if symbol in graph[current_node]:
                current_node = graph[current_node][symbol]
            else:
                # Create a new node
                new_node = next_node
                next_node += 1
                graph[new_node] = {}
                graph[current_node][symbol] = new_node
                
                # Add edge to the result
                trie.append((current_node, new_node, symbol))
                
                # Move to the new node
                current_node = new_node
                
    return trie, graph
````

````python
### Sample Input Execution
sample_input = "ATAGA ATC GAT"
patterns = sample_input.strip().split()

result, _ = TrieConstruction(patterns)

# Format the output
formatted_output = []
for u, v, symbol in result:
    formatted_output.append(f"{u} {v} {symbol}")

# Print the result
for line in formatted_output:
    print(line)

### Verification
expected_output = [
    "0 1 A",
    "1 2 T",
    "2 3 A",
    "3 4 G",
    "4 5 A",
    "2 6 C",
    "0 7 G",
    "7 8 A",
    "8 9 T"
]

assert formatted_output == expected_output, f"Expected {expected_output}, but got {formatted_output}"
print("\nSample test passed!")
````

# Multiple Pattern Matching Problem

```plaintext
PrefixTrieMatching(Text, Trie)
    symbol ← first letter of Text
    v ← root of Trie
    while forever
        if v is a leaf in Trie
            output the pattern spelled by the path from the root to v
        else if there is an edge (v, w) in Trie labeled by symbol
            symbol ← next letter of Text
            v ← w
        else
            return "no matches found"

TrieMatching(Text, Trie)
    while Text is nonempty
        PrefixTrieMatching(Text, Trie)
        remove first symbol from Text
```

**Code Challenge**: Implement TrieMatching to solve the Multiple Pattern Matching Problem.

**Input**: A string $Text$ and a space-separated collection of strings $Patterns$.

**Output**: All starting positions in $Text$ where a string from $Patterns$ appears as a substring.

**Sample Input**:

```
AATCGGGTTCAATCGGGGT
ATCG GGGT
```

**Sample Output**:

```
ATCG: 1 11
GGGT: 4 15
```

````python
def PrefixTrieMatching(text, trie):
    """
    Checks if a prefix of text matches any pattern in the trie.
    Returns the matched pattern if found (at a leaf), else None.
    """
    current_node = 0
    pattern_chars = []
    
    for i, symbol in enumerate(text):
        # Check if current node is a leaf (no outgoing edges)
        if not trie[current_node]: 
            return "".join(pattern_chars)
        
        if symbol in trie[current_node]:
            current_node = trie[current_node][symbol]
            pattern_chars.append(symbol)
        else:
            # No edge matches symbol, and we aren't at a leaf yet
            return None
            
    # If we ran out of text, check if we ended at a leaf
    if not trie[current_node]:
        return "".join(pattern_chars)
        
    return None

def TrieMatching(text, patterns):
    """
    Finds all starting positions in text where a string from patterns appears.
    Returns a dictionary mapping pattern -> list of start indices.
    """
    # Use the graph output from TrieConstruction
    _, trie = TrieConstruction(patterns)
    
    matches = {p: [] for p in patterns}
    
    for i in range(len(text)):
        matched_pattern = PrefixTrieMatching(text[i:], trie)
        if matched_pattern in matches:
            matches[matched_pattern].append(i)
                
    return matches
````

# Maximal Non-branching Paths in a Graph

The $MaximalNonBranchingPaths$ pseudocode below generates all non-branching paths in a graph. It iterates through all nodes of the graph that are not $1$-in-$1$-out nodes and generates all non-branching paths starting at each such node. In a final step, $MaximalNonBranchingPaths$ finds all isolated cycles in the graph.

```plaintext
MaximalNonBranchingPaths(Graph)
    Paths ← empty list
    for each node v in Graph
        if v is not a 1-in-1-out node
            if out(v) > 0
                for each outgoing edge (v, w) from v
                    NonBranchingPath ← the path consisting of single edge (v, w)
                    while w is a 1-in-1-out node
                        extend NonBranchingPath by the edge (w, u) 
                        w ← u
                    add NonBranchingPath to the set Paths
    for each isolated cycle Cycle in Graph
        add Cycle to Paths
    return Paths
```

**Code Challenge**: Implement MaximalNonBranchingPaths.

**Input**: The adjacency list of a graph whose nodes are integers.

**Output**: The collection of all maximal nonbranching paths in this graph.

**Sample Input**:

```
1: 2
2: 3
3: 4 5
6: 7
7: 6
```

**Sample Output**:

```
1 2 3
3 4
3 5
7 6 7
```

````python
def parse_adjacency_list(lines):
    """
    Parses adjacency list text into a dictionary.
    
    Args:
        lines: List of strings, e.g., ["1: 2", "3: 4 5"]
        
    Returns:
        Dictionary {node: [neighbors]}
    """
    graph = {}
    for line in lines:
        if not line.strip():
            continue
        parts = line.strip().split(':')
        if len(parts) >= 2: 
            try:
                node = int(parts[0].strip())
                neighbors = list(map(int, parts[1].strip().split()))
                graph[node] = neighbors
            except ValueError:
                pass
        elif len(parts) == 1 and parts[0].strip():
             try:
                 node = int(parts[0].strip())
                 graph[node] = []
             except ValueError:
                 pass
    return graph

def MaximalNonBranchingPaths(graph):
    """
    Finds all maximal non-branching paths in a graph.
    
    Args:
        graph: A dictionary representing the adjacency list {node: [neighbors]}.
        
    Returns:
        A list of paths, where each path is a list of nodes.
    """
    paths = []
    
    # Get all nodes ensuring we include nodes that only appear as destinations or in values
    all_nodes = set(graph.keys())
    for neighbors in graph.values():
        all_nodes.update(neighbors)
            
    # Calculate in-degrees and out-degrees
    in_degree = {node: 0 for node in all_nodes}
    out_degree = {node: 0 for node in all_nodes}
    
    for u in graph:
        out_degree[u] = len(graph[u])
        for v in graph[u]:
            in_degree[v] += 1
            
    def is_one_in_one_out(v):
        return in_degree[v] == 1 and out_degree[v] == 1

    visited_in_path = set()

    # Step 1: Iterate over all nodes, start paths from non-1-in-1-out nodes
    # We sort to ensure deterministic processing order
    for v in sorted(list(all_nodes)):
        if not is_one_in_one_out(v):
            if out_degree[v] > 0:
                if v in graph:
                    for w in graph[v]:
                        non_branching_path = [v, w]
                        # We don't mark v as visited here in a way that prevents it from being a cycle start 
                        # because v is NOT 1-in-1-out, so it can't be part of an isolated cycle anyway.
                        
                        current = w
                        while is_one_in_one_out(current):
                            visited_in_path.add(current)
                            if current in graph and graph[current]:
                                next_node = graph[current][0]
                                non_branching_path.append(next_node)
                                current = next_node
                            else:
                                break
                        
                        paths.append(non_branching_path)

    # Step 2: Find isolated cycles
    visited_in_cycle = set() 
    
    for v in sorted(list(all_nodes)):
        # Must be 1-in-1-out, and not part of a previously generated path
        if is_one_in_one_out(v) and v not in visited_in_path and v not in visited_in_cycle:
            cycle = [v]
            visited_in_cycle.add(v)
            current = v
            
            while True:
                if current in graph and graph[current]:
                    next_node = graph[current][0]
                    
                    if next_node == v:
                        # Cycle detected and closed
                        cycle.append(next_node)
                        paths.append(cycle)
                        break
                    
                    if next_node in visited_in_cycle or next_node in visited_in_path:
                        # Should not happen for valid isolated cycles
                        break
                        
                    visited_in_cycle.add(next_node)
                    cycle.append(next_node)
                    current = next_node
                else:
                    break
                    
    return paths
````

````python
### Sample Input Execution
sample_input = """
1: 2
2: 3
3: 4 5
6: 7
7: 6
""".strip().split('\n')

graph = parse_adjacency_list(sample_input)
paths = MaximalNonBranchingPaths(graph)

formatted_output = []
for path in paths:
    path_str = " ".join(map(str, path))
    formatted_output.append(path_str)
    print(path_str)

### Verification
# Note: For the cycle 6-7, both "6 7 6" and "7 6 7" are valid representations of the isolated cycle.
# The algorithm output order for the cycle starts depends on iteration order.
# We accept either for verification.

expected_paths = {
    "1 2 3",
    "3 4",
    "3 5"
}
expected_cycles = {"6 7 6", "7 6 7"}

output_set = set(formatted_output)

# Check standard paths
missing_paths = expected_paths - output_set
assert not missing_paths, f"Missing expected paths: {missing_paths}"

# Check cycle
found_cycle = output_set.intersection(expected_cycles)
assert found_cycle, f"Missing expected cycle. Got: {output_set}"

print("\nSample test passed!")
````

# Suffix Trees

The following pseudocode constructs the modified suffix trie of a string $Text$ by traversing the suffixes of $Text$ from longest to shortest. Given a suffix, it attempts to spell the suffix by moving downward in the tree, following edge labels as far as possible until it can go no further. At that point, it adds the rest of the suffix to the trie in the form of a path to a leaf, along with the position of each symbol in the suffix.

```plaintext
ModifiedSuffixTrieConstruction(Text)
    Trie ← a graph consisting of a single node root
    for i ← 0 to |Text| - 1
        currentNode ← root
        for j ← i to |Text| - 1
            currentSymbol ← j-th symbol of Text
            if there is an outgoing edge from currentNode labeled by currentSymbol
                currentNode ← ending node of this edge
            else
                add a new node newNode to Trie
                add an edge newEdge connecting currentNode to newNode in Trie
                Symbol(newEdge) ← currentSymbol
                Position(newEdge) ← j
                currentNode ← newNode
        if currentNode is a leaf in Trie
            assign label i to this leaf
    return Trie

ModifiedSuffixTreeConstruction(Text)
    Trie ← ModifiedSuffixTrieConstruction
    for each non-branching path Path in Trie
        substitute Path by a single edge e connecting the first and last nodes of Path
        Position(e) ← Position(first edge of Path)
        Length(e) ← number of edges of Path
    return Trie
```

**Code Challenge**: Solve the Suffix Tree Construction Problem.

**Input**: A string $Text$.

**Output**: A space-separated list of the edge labels of $SuffixTree(Text)$. You may return these strings in any order.

**Sample Input**:

```
ATAAATG$
```

**Sample Output**:

```
AAATG$ G$ T ATG$ TG$ A A AAATG$ G$ T G$ $
```

````python
def SuffixTreeConstruction(text):
    """
    Constructs a Suffix Tree for the text.
    
    Args:
        text: Input string.
        
    Returns:
        A tuple containing:
        1. List of strings representing the edge labels.
        2. The root dictionary of the tree structure.
    """
    root = {} # key: first char of label, value: (child_dict, full_label_string)
    
    for i in range(len(text)):
        suffix = text[i:]
        current_node = root
        j = 0
        
        while j < len(suffix):
            char = suffix[j]
            
            if char in current_node:
                child_node, label = current_node[char]
                
                # Find the length of the matching prefix
                match_len = 0
                while match_len < len(label) and j + match_len < len(suffix) and label[match_len] == suffix[j + match_len]:
                    match_len += 1
                    
                if match_len == len(label):
                    # Full match of the edge label, proceed to the child node
                    current_node = child_node
                    j += match_len
                else:
                    # Partial match - we need to split the edge
                    split_node = {}
                    
                    # Update current_node to point to split_node with the matched prefix part of the label
                    current_node[char] = (split_node, label[:match_len])
                    
                    # The split_node points to the original child with the remaining part of the original label
                    remaining_label = label[match_len:]
                    split_node[remaining_label[0]] = (child_node, remaining_label)
                    
                    # The split_node also points to a new leaf with the rest of the current suffix
                    remaining_suffix = suffix[j + match_len:]
                    new_leaf = {}
                    split_node[remaining_suffix[0]] = (new_leaf, remaining_suffix)
                    
                    # Suffix insertion complete
                    break
            else:
                # No edge starts with this character, create a new edge to a new leaf
                new_leaf = {}
                current_node[char] = (new_leaf, suffix[j:])
                break
                
    # Collect all edges
    edges = []
    stack = [root]
    while stack:
        node = stack.pop()
        for char in node:
            child, label = node[char]
            edges.append(label)
            stack.append(child)
            
    return edges, root
````

# Longest Repeat Problem

Find the longest repeat in a string.

**Input**: A string $Text$.

**Output**: A longest substring of $Text$ that appears in $Text$ more than once.

**Code Challenge**: Solve the Longest Repeat Problem. (Multiple solutions may exist, in which case you may return any one.)

**Sample Input**:

```
ATATCGTTTTATCGTT
```

**Sample Output**:

```
TATCGTT
```

````python
import sys

# Increasing recursion depth just in case, though we use iterative approach mostly
sys.setrecursionlimit(2000)

def LongestRepeat(text):
    """
    Finds the longest repeat in a string using a suffix tree.
    
    Args:
        text: Input string.
        
    Returns:
        The longest substring that appears in text more than once.
    """
    # Append unique terminator to ensure proper suffix tree structure
    text_with_term = text + "$"
    # Reuse the modified SuffixTreeConstruction
    _, tree_root = SuffixTreeConstruction(text_with_term)
    
    longest_substring = ""
    
    # DFS to find the deepest internal node (deepest by string length)
    # Stack stores: (node_dict, current_path_string)
    stack = [(tree_root, "")]
    
    while stack:
        current_node, current_path = stack.pop()
        
        # Check if it is an internal node
        # A node is internal if it has children. 
        # In our construction, leaves are empty dicts {}.
        is_internal = bool(current_node)
        
        if is_internal:
            if len(current_path) > len(longest_substring):
                longest_substring = current_path
        
            for char in current_node:
                child_node, label = current_node[char]
                
                # We only push valid continuations.
                stack.append((child_node, current_path + label))
                
    return longest_substring
````

# Longest Shared Substring Problem

Find the longest substring shared by two strings.

**Input**: Strings $Text1$ and $Text2$.

**Output**: The longest substring that occurs in both $Text1$ and $Text2$.

**Code Challenge**: Solve the Longest Shared Substring Problem. (Multiple solutions may exist, in which case you may return any one.)

**Sample Input**:

```
TCGGTAGATTGCGCCCACTC
AGGGGCTCGCAGTGTAAGAA
```

**Sample Output**:

```
AGA
```

````python
def LongestSharedSubstring(text1, text2):
    """
    Finds the longest substring shared by two strings using a Suffix Tree.
    
    Args:
        text1: First string.
        text2: Second string.
        
    Returns:
        The longest shared substring.
    """
    # Use unique terminators
    term1 = "#"
    term2 = "$"
    combined_text = text1 + term1 + text2 + term2
    
    # Construct generalized suffix tree
    _, tree_root = SuffixTreeConstruction(combined_text)
    
    max_len = 0
    longest_substring = ""
    
    def traverse(node, current_path):
        nonlocal max_len, longest_substring
        
        is_t1_descendant = False
        is_t2_descendant = False
        
        for char in node:
            child, label = node[char]
            
            if term1 in label:
                # Path contains #, so it MUST be from Text1 part
                child_is_t1 = True
                child_is_t2 = False
            elif term2 in label:
                # Path contains $ (and not #, per previous check), so it MUST be from Text2 part
                child_is_t1 = False
                child_is_t2 = True
            else:
                # No terminators in this edge, recurse deeper
                c_t1, c_t2 = traverse(child, current_path + label)
                child_is_t1 = c_t1
                child_is_t2 = c_t2
            
            is_t1_descendant = is_t1_descendant or child_is_t1
            is_t2_descendant = is_t2_descendant or child_is_t2
            
        # If this node has descendants from both Text1 and Text2, it represents a shared substring
        if is_t1_descendant and is_t2_descendant:
            if len(current_path) > max_len:
                max_len = len(current_path)
                longest_substring = current_path
                
        return is_t1_descendant, is_t2_descendant

    traverse(tree_root, "")
    return longest_substring
````

````python
### Sample Input Execution
sample_input = """
TCGGTAGATTGCGCCCACTC
AGGGGCTCGCAGTGTAAGAA
""".strip().split('\n')

text1_sample = sample_input[0].strip()
text2_sample = sample_input[1].strip()

result = LongestSharedSubstring(text1_sample, text2_sample)
print(result)

### Verification
expected_output = "AGA"
assert result in text1_sample and result in text2_sample, f"Result {result} not in inputs"
print(f"Result '{result}' is a valid shared substring.")

if result == expected_output:
    print("Matches sample output exactly.")
else:
    print(f"Sample output {expected_output}, got {result}. both valid if same length.")

print("\nSample test passed!")
````

# Shortest Non-Shared Substring Problem

Find the shortest substring of one string that does not appear in another string.

**Input**: Strings $Text1$ and $Text2$.

**Output**: The shortest substring of $Text1$ that does not appear in $Text2$.

**Code Challenge**: Solve the Shortest Non-Shared Substring Problem. (Multiple solutions may exist, in which case you may return any one.)

**Sample Input**:

```
CCAAGCTGCTAGAGG
CATGCTGGGCTGGCT
```

**Sample Output**:

```
AA
```

````python
def ShortestNonSharedSubstring(text1, text2):
    """
    Finds the shortest substring of text1 that does not appear in text2.
    
    Args:
        text1: The string to search for a non-shared substring.
        text2: The string to check against.
        
    Returns:
        The shortest substring of text1 that is not in text2.
    """
    # Build Suffix Tree for Text2 with a terminator
    # This tree contains all substrings of Text2 as prefixes of suffixes
    _, tree_root = SuffixTreeConstruction(text2 + "#")
    
    min_len = float('inf')
    result = ""
    
    # Iterate through all suffixes of Text1
    for i in range(len(text1)):
        suffix = text1[i:]
        
        current_node = tree_root
        j = 0 # Index in the current suffix
        
        while j < len(suffix):
            char = suffix[j]
            
            if char in current_node:
                child_node, label = current_node[char]
                
                # Try to traverse the edge
                k = 0
                match_edge = True
                while k < len(label) and j < len(suffix):
                    if label[k] != suffix[j]:
                        match_edge = False
                        break
                    k += 1
                    j += 1
                
                if not match_edge:
                    # Mismatch found within the edge
                    # The substring ending at j is the first non-matching char
                    # Length is j+1 relative to the start of this specific suffix trace
                    # The length of this non-shared substring is j (0-based index) + 1
                    # Since j has been incremented to the mismatch position.
                    candidate_len = j + 1
                    if candidate_len < min_len:
                        min_len = candidate_len
                        result = text1[i : i + candidate_len]
                    break
                
                # If we successfully traversed the edge, move to the child
                current_node = child_node
                
            else:
                # Immediate mismatch (no edge starting with this char)
                # The first character itself is non-shared at this point in the path
                candidate_len = j + 1
                if candidate_len < min_len:
                    min_len = candidate_len
                    result = text1[i : i + candidate_len]
                break
                
    return result
````

````python
### Sample Input Execution
sample_input = """
CCAAGCTGCTAGAGG
CATGCTGGGCTGGCT
""".strip().split('\n')

text1_sample = sample_input[0].strip()
text2_sample = sample_input[1].strip()

result = ShortestNonSharedSubstring(text1_sample, text2_sample)
print(result)

### Verification
# "AA" and "CC" are both valid shortest non-shared substrings.
expected_output = "CC" 
assert result == expected_output, f"Expected {expected_output}, but got {result}"
print("\nSample test passed!")
````

# Tree Coloring Problem

Color the internal nodes of a tree given the colors of its leaves.

**Input**: An adjacency list, followed by color labels for leaf nodes.

**Output**: Color labels for all nodes, in any order.

**Code Challenge**: Solve the Tree Coloring Problem.

**Sample Input**:

```
0:
1:
2: 0 1
3:
4:
5: 2 3
6:
7: 4 5 6
-
0 red
1 red
3 blue
4 blue
6 red
```

**Sample Output**:

```
0 red
1 red
2 red
3 blue
4 blue
5 purple
6 red
7 purple
```

````python
def parse_tree_coloring_input(lines):
    """
    Parses the tree coloring input format.
    
    Args:
        lines: List of strings from input.
        
    Returns:
        adj: Adjacency list dictionary.
        colors: Dictionary of initial node colors.
    """
    adj_lines = []
    color_lines = []
    
    separator_found = False
    for line in lines:
        line = line.strip()
        if line == '-':
            separator_found = True
            continue
            
        if not separator_found:
            adj_lines.append(line)
        else:
            color_lines.append(line)
            
    # Parse Adjacency
    adj = parse_adjacency_list(adj_lines)
    
    # Parse Colors
    leaf_colors = {}
    for line in color_lines:
        parts = line.split()
        if len(parts) >= 2:
            try:
                # Remove colon if present in "0: red"
                node_str = parts[0].strip().replace(':', '')
                node = int(node_str)
                color = parts[1].strip()
                leaf_colors[node] = color
            except ValueError:
                pass
                
    return adj, leaf_colors

def TreeColoring(adj, colors):
    """
    Colors the internal nodes of a tree based on leaf colors.
    
    Args:
        adj: Adjacency dictionary {node: [children]}
        colors: Dictionary {node: color_string} for leaves
        
    Returns:
        Dictionary {node: color_string} for all nodes
    """
    # Identify all nodes
    all_nodes = set(adj.keys())
    for children in adj.values():
        all_nodes.update(children)
        
    final_colors = colors.copy()
    
    # Memoized color function
    memo = {}
    
    def get_color(u):
        if u in final_colors:
            return final_colors[u]
        
        if u in memo:
            return memo[u]
        
        if u in adj and adj[u]:
            child_colors = [get_color(v) for v in adj[u]]
            
            # Logic: 
            # If all children have the same color, parent takes that color.
            # Else, parent is "purple".
            
            distinct = set(child_colors)
            if len(distinct) == 1:
                res = distinct.pop()
            else:
                res = "purple"
                
            final_colors[u] = res
            memo[u] = res
            return res
        else:
            # It's a leaf but wasn't in initial colors.
            # Should not happen per problem statement, or maybe is uncolored?
            # We return a placeholder or handle gracefully.
            return ValueError("Leaf node without initial color.")

    # Process all nodes
    # We sort to ensure deterministic iteration, though recursion handles dependency
    for node in sorted(list(all_nodes)):
        get_color(node)
        
    return final_colors
````

````python
### Sample Input Execution
sample_input = """
0:
1:
2: 0 1
3:
4:
5: 2 3
6:
7: 4 5 6
-
0 red
1 red
3 blue
4 blue
6 red
""".strip().split('\n')

adj, leaf_colors = parse_tree_coloring_input(sample_input)
final_colors = TreeColoring(adj, leaf_colors)

formatted_output = []
for node in sorted(final_colors.keys()):
    line = f"{node} {final_colors[node]}"
    formatted_output.append(line)
    print(line)

### Verification
expected_output = [
    "0 red",
    "1 red",
    "2 red",
    "3 blue",
    "4 blue",
    "5 purple",
    "6 red",
    "7 purple"
]

assert formatted_output == expected_output, f"Expected {expected_output}, but got {formatted_output}"
print("\nSample test passed!")
````

````python
# Coursera Quiz Questions

# Q5 How many leaves will SuffixTree("TTGAATGACTCCTATAACGAACTTCGACATGGCA$") have?
def count_suffix_tree_leaves(text):
    _, tree_root = SuffixTreeConstruction(text)
    
    # Count leaves in the suffix tree
    def count_leaves(node):
        if not node:
            return 1  # Leaf node
        total = 0
        for char in node:
            child_node, _ = node[char]
            total += count_leaves(child_node)
        return total
    
    return count_leaves(tree_root)

print("Q5:", count_suffix_tree_leaves("TTGAATGACTCCTATAACGAACTTCGACATGGCA$"))
````
