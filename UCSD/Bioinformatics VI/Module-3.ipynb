# Better BWMatching

We can match patterns in string $Text$ more efficiently using the Burrows-Wheeler Transform.

```plaintext
BetterBWMatching(FirstOccurrence, LastColumn, Pattern, Count)
    top ← 0
    bottom ← |LastColumn| − 1
    while top ≤ bottom
        if Pattern is nonempty
            symbol ← last letter in Pattern
            remove last letter from Pattern
            if positions from top to bottom in LastColumn contain an occurrence of symbol
                top ← FirstOccurrence(symbol) + Countsymbol(top, LastColumn)
                bottom ← FirstOccurrence(symbol) + Countsymbol(bottom + 1, LastColumn) − 1
            else
                return 0
        else
            return bottom − top + 1
```

**Code Challenge**: Implement BetterBWMatching.

**Input**: A string $BWT(Text)$ followed by a space-separated collection of strings $Patterns$.

**Output**: A space-separated list of integers, where the $i$-th integer corresponds to the number of substring matches of the $i$-th member of $Patterns$ in $Text$.

**Sample Input**:

```
GGCGCCGC$TAGTCACACACGCCGTA
ACC CCG CAG
```

**Sample Output**:

```
1 2 1
```

````python
def precompute_bwt_data(bwt):
    """
    Precomputes the FirstOccurrence and Count matrices for BetterBWMatching.
    
    Args:
    bwt (str): The Burrows-Wheeler Transform of the text.
    
    Returns:
    dict, dict: FirstOccurrence map and Count matrix.
    """
    sorted_bwt = sorted(bwt)
    first_occurrence = {}
    for i, char in enumerate(sorted_bwt):
        if char not in first_occurrence:
            first_occurrence[char] = i
            
    # Count matrix: count_dict[char][i] = occurrences of char in bwt[:i]
    unique_chars = set(bwt)
    count_dict = {char: [0] * (len(bwt) + 1) for char in unique_chars}
    
    for i in range(len(bwt)):
        current_char = bwt[i]
        for char in unique_chars:
            count_dict[char][i+1] = count_dict[char][i]
        count_dict[current_char][i+1] += 1
        
    return first_occurrence, count_dict

def BetterBWMatching(first_occurrence, bwt, pattern, count_dict):
    """
    Finds the number of occurrences of a pattern in the text using Better BWMatching.
    
    Args:
    first_occurrence (dict): Map of first occurrence indices in sorted first column.
    bwt (str): The Burrows-Wheeler Transform of the text.
    pattern (str): The pattern to search for.
    count_dict (dict): The Count matrix.
    
    Returns:
    int: The number of occurrences of the pattern.
    """
    top = 0
    bottom = len(bwt) - 1
    
    while top <= bottom:
        if pattern:
            symbol = pattern[-1]
            pattern = pattern[:-1]
            
            if symbol in count_dict:
                # Check if occurrences exist in the current range
                if count_dict[symbol][bottom + 1] - count_dict[symbol][top] > 0:
                    top = first_occurrence[symbol] + count_dict[symbol][top]
                    bottom = first_occurrence[symbol] + count_dict[symbol][bottom + 1] - 1
                else:
                    return 0
            else:
                return 0
        else:
            return bottom - top + 1
    return 0

def solve_better_bw_matching(bwt, patterns):
    """
    Solves the Better BWMatching problem for a list of patterns.
    """
    first_occurrence, count_dict = precompute_bwt_data(bwt)
    results = []
    
    for pattern in patterns:
        results.append(BetterBWMatching(first_occurrence, bwt, pattern, count_dict))
        
    return results
````

````python
### Sample Input Execution
sample_bwt = "GGCGCCGC$TAGTCACACACGCCGTA"
sample_patterns = "ACC CCG CAG".split()

results = solve_better_bw_matching(sample_bwt, sample_patterns)

# Print output
print(" ".join(map(str, results)))

### Verification
expected_output = [1, 2, 1]

assert results == expected_output, f"Expected {expected_output}, but got {results}"
print("Sample test passed!")
````

# Multiple Pattern Matching Problem

**Code Challenge**: Solve the Multiple Pattern Matching Problem.

**Input**: A string $Text$ followed by a space-separated collection of strings $Patterns$.

**Output**: For each string $Pattern$ in $Patterns$, the string $Pattern$ followed by a colon, followed by a space-separated list of all starting positions in $Text$ where $Pattern$ appears as a substring.

**Sample Input**:

```
AATCGGGTTCAATCGGGGT
ATCG GGGT
```

**Sample Output**:

```
ATCG: 1 11
GGGT: 4 15
```

````python
def parse_multiple_pattern_matching_input(lines):
    lines = [l.strip() for l in lines if l.strip()]
    if not lines:
        return "", []
    text = lines[0]
    patterns = lines[1].split() if len(lines) > 1 else []
    return text, patterns

def extract_suffix_array(text):
    """
    Constructs the suffix array of a string.
    """
    suffixes = []
    for i in range(len(text)):
        suffixes.append((text[i:], i))
    suffixes.sort()
    suffix_array = [s[1] for s in suffixes]
    return suffix_array

def bwt_from_suffix_array(text, suffix_array):
    """
    Constructs BWT from the Suffix Array.
    """
    n = len(text)
    bwt = [''] * n
    for i in range(n):
        idx = suffix_array[i]
        bwt[i] = text[(idx - 1) % n]
    return "".join(bwt)

def BWMatchingRange(first_occurrence, bwt, pattern, count_dict):
    """
    Finds the range [top, bottom] of suffixes starting with pattern.
    """
    top = 0
    bottom = len(bwt) - 1
    
    while top <= bottom:
        if pattern:
            symbol = pattern[-1]
            pattern = pattern[:-1]
            
            if symbol in count_dict:
                if count_dict[symbol][bottom + 1] - count_dict[symbol][top] > 0:
                    top = first_occurrence[symbol] + count_dict[symbol][top]
                    bottom = first_occurrence[symbol] + count_dict[symbol][bottom + 1] - 1
                else:
                    return None
            else:
                return None
        else:
            return (top, bottom)
    return None

def solve_multiple_pattern_matching(text, patterns):
    # 1. Append terminator
    text_with_term = text + "$"
    
    # 2. Build Suffix Array and BWT
    suffix_array = extract_suffix_array(text_with_term)
    bwt = bwt_from_suffix_array(text_with_term, suffix_array)
    
    # 3. Precompute tables
    first_occurrence, count_dict = precompute_bwt_data(bwt)
    
    results = {}
    
    for pattern in patterns:
        match_range = BWMatchingRange(first_occurrence, bwt, pattern, count_dict)
        if match_range:
            top, bottom = match_range
            # Map back to original indices using suffix_array
            # Suffix Array indices correspond to the start of the suffix in 'text_with_term'
            indices = sorted([suffix_array[i] for i in range(top, bottom + 1)])
            results[pattern] = indices
        else:
            results[pattern] = []
            
    return results
````

````python
### Sample Input Execution
sample_input = """
AATCGGGTTCAATCGGGGT
ATCG GGGT
""".strip().split('\n')

text_sample, patterns_sample = parse_multiple_pattern_matching_input(sample_input)
results = solve_multiple_pattern_matching(text_sample, patterns_sample)

for pattern in patterns_sample:
    if pattern in results:
        positions = results[pattern]
        print(f"{pattern}: {' '.join(map(str, positions))}")

### Verification
expected_output = {
    "ATCG": [1, 11],
    "GGGT": [4, 15]
}

for pattern, expected in expected_output.items():
    assert results[pattern] == expected, f"For {pattern}, expected {expected}, got {results[pattern]}"
print("Sample test passed!")
````

# Multiple Approximate Pattern Matching Problem

**Code Challenge**: Solve the Multiple Approximate Pattern Matching Problem.

**Input**: A string $Text$, followed by a collection of space-separated strings $Patterns$, and an integer $d$.

**Output**: For each string $Pattern$ in $Patterns$, the string $Pattern$ followed by a colon, followed by a space-separated collection of all positions where $Pattern$ appears as a substring of $Text$ with at most $d$ mismatches.

**Sample Input**:

```
ACATGCTACTTT
ATT GCC GCTA TATT
1
```

**Sample Output**:

```
ATT: 2 7 8 9
GCC: 4
GCTA: 4
TATT: 6
```

````python
def parse_approximate_pattern_input(lines):
    # Filter empty lines
    lines = [l.strip() for l in lines if l.strip()]
    if len(lines) < 3:
        return "", [], 0
    text = lines[0]
    patterns = lines[1].split()
    d = int(lines[2])
    return text, patterns, d

def solve_multiple_approximate_pattern_matching(text, patterns, d):
    # 1. Build BWT
    text_with_term = text + "$"
    suffix_array = extract_suffix_array(text_with_term)
    bwt = bwt_from_suffix_array(text_with_term, suffix_array)
    
    # 2. Precompute FirstOccurrence and Count matrices
    first_occurrence, count_dict = precompute_bwt_data(bwt)
    
    results = {}
    
    # Recursive function to find approximate matches
    def recursive_match(top, bottom, p_idx, mismatches, pattern, matches_set):
        if mismatches > d:
            return
        
        if p_idx < 0:
            # Pattern matched within tolerance. Add all occurences in this range.
            for i in range(top, bottom + 1):
                # Map BWT range index back to original text index via Suffix Array
                matches_set.add(suffix_array[i])
            return

        current_char = pattern[p_idx]
        
        # Iterate over all possible characters in the BWT (alphabet of Text)
        for char, f_start in first_occurrence.items():
            if char == '$': 
                continue # Cannot match $ in middle of pattern
            
            # Get count of char in current [top, bottom] range
            c_top = count_dict[char][top]
            c_bottom = count_dict[char][bottom + 1]
            
            occurrences = c_bottom - c_top
            
            if occurrences > 0:
                # Update range using FM-Index property
                new_top = f_start + c_top
                new_bottom = f_start + c_bottom - 1
                
                # Calculate cost
                cost = 0 if char == current_char else 1
                
                recursive_match(new_top, new_bottom, p_idx - 1, mismatches + cost, pattern, matches_set)
    
    for pattern in patterns:
        matched_indices = set()
        recursive_match(0, len(bwt) - 1, len(pattern) - 1, 0, pattern, matched_indices)
        results[pattern] = sorted(list(matched_indices))
        
    return results
````

````python
### Sample Input Execution
sample_input = """
ACATGCTACTTT
ATT GCC GCTA TATT
1
""".strip().split('\n')

text_s, patterns_s, d_s = parse_approximate_pattern_input(sample_input)
results_s = solve_multiple_approximate_pattern_matching(text_s, patterns_s, d_s)

# Start Output
for pattern in patterns_s:
    if pattern in results_s:
        # Format: Pattern: idx1 idx2 ...
        print(f"{pattern}: {' '.join(map(str, results_s[pattern]))}")

### Verification
expected_output = {
    "ATT": [2, 7, 8, 9],
    "GCC": [4],
    "GCTA": [4],
    "TATT": [6]
}

for pattern, expected in expected_output.items():
    assert results_s[pattern] == expected, f"For {pattern}, expected {expected}, got {results_s[pattern]}"
print("Sample test passed!")
````

````python
def inverse_bwt(bwt):
    # Map each character in the last column (bwt) to its position in the first column (sorted bwt)
    last_col = [(bwt[i], i) for i in range(len(bwt))]
    first_col = sorted(last_col)
    
    # t_map[i] maps the i-th row in the matrix (determined by bwt[i]) 
    # to the row index in the first column corresponding to the same character instance.
    t_map = [0] * len(bwt)
    for j, (char, orig_idx) in enumerate(first_col):
        t_map[orig_idx] = j
        
    text = ""
    curr = 0 # Row starting with '$' is always first in sorted suffixes
    for _ in range(len(bwt)):
        char = bwt[curr]
        text = char + text
        curr = t_map[curr]
    
    # The reconstructed text will start with '$' due to the backward reconstruction logic
    # touching the cyclic end. Move '$' to the end.
    if text.startswith('$'):
        text = text[1:] + '$'
    return text
````
