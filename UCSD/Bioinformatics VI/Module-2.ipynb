# Suffix Array Construction Problem

Construct the suffix array of a string.

**Code Challenge**: Solve the Suffix Array Construction Problem.

**Input**: A string $Text$.

**Output**: $SuffixArray(Text)$, as a space-separated collection of integers.

**Sample Input**:

```
AACGATAGCGGTAGA$
```

**Sample Output**:

```
15 14 0 1 12 6 4 2 8 13 3 7 9 10 11 5
```

````python
def SuffixArray(text):
    """
    Constructs the suffix array of a string.
    
    Args:
    text (str): The input string.
    
    Returns:
    list: A list of integers representing the starting indices of the sorted suffixes.
    """
    suffixes = []
    n = len(text)
    for i in range(n):
        # Store suffix and its original index
        suffixes.append((text[i:], i))
    
    # Sort suffixes lexicographically
    suffixes.sort()
    
    # Extract the original indices
    suffix_array = [idx for _, idx in suffixes]
    return suffix_array
````

````python
### Sample Input Execution
sample_input = "AACGATAGCGGTAGA$"
result = SuffixArray(sample_input)

# Print formatted output
print(" ".join(map(str, result)))

### Verification
expected_output = [15, 14, 0, 1, 12, 6, 4, 2, 8, 13, 3, 7, 9, 10, 11, 5]

assert result == expected_output, f"Expected {expected_output}, but got {result}"
print("Sample test passed!")
````

# Burrows-Wheeler Transform Construction Problem

Construct the Burrows-Wheeler transform of a string.

**Code Challenge**: Solve the Burrows-Wheeler Transform Construction Problem.

**Input**: A string $Text$.

**Output**: $BWT(Text)$.

Note: Although it is possible to construct the Burrows-Wheeler transform in $O(|Text|)$ time and space, we do not expect you to implement such a fast algorithm. In other words, it is perfectly fine to produce $BWT(Text)$ by first producing the complete Burrows-Wheeler matrix $M(Text)$.

**Sample Input**:

```
GCGTGCCTGGTCA$
```

**Sample Output**:

```
ACTGGCT$TGCGGC
```

````python
def BurrowsWheelerTransform(text):
    """
    Constructs the Burrows-Wheeler Transform of a string.
    
    Args:
    text (str): The input string.
    
    Returns:
    str: The Burrows-Wheeler Transform of the string.
    """
    n = len(text)
    suffix_array = SuffixArray(text)
    
    # BWT[i] = Text[(SA[i] - 1) % n]
    bwt = []
    for i in range(n):
        bwt.append(text[(suffix_array[i] - 1) % n])
        
    return "".join(bwt)
````

````python
### Sample Input Execution
sample_input = "GCGTGCCTGGTCA$"
result = BurrowsWheelerTransform(sample_input)

# Print output
print(result)

### Verification
expected_output = "ACTGGCT$TGCGGC"

assert result == expected_output, f"Expected {expected_output}, but got {result}"
print("Sample test passed!")
````

# Inverse Burrows-Wheeler Transform Problem

Reconstruct a string from its Burrows-Wheeler transform.

**Code Challenge**: Solve the Inverse Burrows-Wheeler Transform Problem.

**Input**: A string $Transform$ (with a single "$" symbol).

**Output**: The string $Text$ such that $BWT(Text) = Transform$.

**Sample Input**:

```
TTCCTAACG$A
```

**Sample Output**:

```
TACATCACGT$
```

````python
def InverseBurrowsWheelerTransform(bwt):
    """
    Reconstructs the original string from its Burrows-Wheeler Transform.
    
    Args:
    bwt (str): The Burrows-Wheeler Transform of a string.
    
    Returns:
    str: The original string.
    """
    n = len(bwt)
    
    # Store tuples of (character, original_index)
    last_col = [(bwt[i], i) for i in range(n)]
    
    # Sort to get the first column
    first_col = sorted(last_col)
    
    # Create the LF mapping
    # Maps index in last column to corresponding index in first column
    lf_map = [0] * n
    for j, (char, original_index) in enumerate(first_col):
        lf_map[original_index] = j
        
    # Reconstruct the string
    reconstructed = []
    current_idx = 0  # Start with the row beginning with '$' (lexicographically first)
    
    for _ in range(n):
        reconstructed.append(bwt[current_idx])
        current_idx = lf_map[current_idx]
        
    # Reverse to get proper order and move '$' to the end
    return "".join(reconstructed[::-1])[1:] + "$"
````

````python
### Sample Input Execution
sample_input = "TTCCTAACG$A"
result = InverseBurrowsWheelerTransform(sample_input)

# Print output
print(result)

### Verification
expected_output = "TACATCACGT$"

assert result == expected_output, f"Expected {expected_output}, but got {result}"
print("Sample test passed!")
````

# Better BWMatching Algorithm

```plaintext
BWMatching(LastColumn, Pattern, LastToFirst)
    top ← 0
    bottom ← |LastColumn| − 1
    while top ≤ bottom
        if Pattern is nonempty
            symbol ← last letter in Pattern
            remove last letter from Pattern
            if positions from top to bottom in LastColumn contain an occurrence of symbol
                topIndex ← first position of symbol among positions from top to bottom in LastColumn
                bottomIndex ← last position of symbol among positions from top to bottom in LastColumn
                top ← LastToFirst(topIndex)
                bottom ← LastToFirst(bottomIndex)
            else
                return 0
        else
            return bottom − top + 1
```

**Code Challenge**: Implement BWMatching.

**Input**: A string $BWT(Text)$, followed by a space-separated collection of $Patterns$.

**Output**: A space-separated list of integers, where the $i$-th integer corresponds to the number of substring matches of the $i$-th member of $Patterns$ in $Text$.

**Sample Input**:

```
TCCTCTATGAGATCCTATTCTATGAAACCTTCA$GACCAAAATTCTCCGGC
CCT CAC GAG CAG ATC
```

**Sample Output**:

```
2 1 1 0 1
```

````python
def compute_last_to_first(bwt):
    """
    Computes the Last-to-First mapping for the BWT.
    """
    n = len(bwt)
    last_col = [(bwt[i], i) for i in range(n)]
    first_col = sorted(last_col)
    
    lf_map = [0] * n
    for j, (char, original_index) in enumerate(first_col):
        lf_map[original_index] = j
    return lf_map

def BWMatching(last_column, pattern, last_to_first):
    """
    Finds the number of occurrences of a pattern in the text using BWT.
    
    Args:
    last_column (str): The Burrows-Wheeler Transform of the text.
    pattern (str): The pattern to search for.
    last_to_first (list): The Last-to-First mapping.
    
    Returns:
    int: The number of occurrences of the pattern.
    """
    top = 0
    bottom = len(last_column) - 1
    while top <= bottom:
        if pattern:
            symbol = pattern[-1]
            pattern = pattern[:-1]
            
            # Check for occurrence of symbol in last_column[top:bottom+1]
            search_scope = last_column[top : bottom + 1]
            first_occ = search_scope.find(symbol)
            
            if first_occ != -1:
                last_occ = search_scope.rfind(symbol)
                topIndex = top + first_occ
                bottomIndex = top + last_occ
                
                top = last_to_first[topIndex]
                bottom = last_to_first[bottomIndex]
            else:
                return 0
        else:
            return bottom - top + 1
    return 0

def solve_bw_matching_problem(bwt, patterns):
    """
    Wrapper to solve for multiple patterns.
    """
    last_to_first = compute_last_to_first(bwt)
    results = []
    
    for pattern in patterns:
        results.append(BWMatching(bwt, pattern, last_to_first))
        
    return results
````

````python
### Sample Input Execution
sample_bwt = "TCCTCTATGAGATCCTATTCTATGAAACCTTCA$GACCAAAATTCTCCGGC"
sample_patterns = "CCT CAC GAG CAG ATC".split()

results = solve_bw_matching_problem(sample_bwt, sample_patterns)

# Print output
print(" ".join(map(str, results)))

### Verification
expected_output = [2, 1, 1, 0, 1]

assert results == expected_output, f"Expected {expected_output}, but got {results}"
print("Sample test passed!")
````
