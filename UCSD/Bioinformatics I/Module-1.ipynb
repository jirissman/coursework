```python
def PatternCount(text, pattern):
    count = 0
    for i in range(len(text) - len(pattern) + 1):
        if text[i:i+len(pattern)] == pattern:
            count += 1
    return count
```

```python
import os

input_dir = 'PatternCount/inputs'
output_dir = 'PatternCount/myoutputs'

for i in range(1, 7):
    input_file = os.path.join(input_dir, f'input_{i}.txt')
    output_file = os.path.join(output_dir, f'output_{i}.txt')
    
    with open(input_file, 'r') as f:
        text = f.readline().strip()
        pattern = f.readline().strip()
        result = PatternCount(text, pattern)
    
    with open(output_file, 'x') as f:
        f.write(str(result))
    
    print(f"Processed {input_file} -> {output_file}: {result}")
```

```python
from collections import defaultdict

def FrequencyTable(text, k):
    freqMap = defaultdict(int)
    for i in range(len(text)-k+1):
        freqMap[text[i:i+k]] += 1
    return dict(freqMap)

def FrequentWords(text, k):
    freqMap = FrequencyTable(text, k)
    max_freq = max(freqMap.values())
    return [pattern for pattern, freq in freqMap.items() if freq == max_freq]
    
```

```python
def PatternMatch(pattern, genome):
    positions = []
    m, n = len(pattern), len(genome)
    if m == 0 or m > n:
        return positions
    for i in range(n - m + 1):
        if genome[i:i+m] == pattern:
            positions.append(i)
    return positions
```

```python
from collections import defaultdict

def FindClumps(Text, k, L, t):
    Patterns = set()
    n = len(Text)
    
    # Initialize frequency map for the first window using FrequencyTable
    window = Text[0:L]
    freqMap = defaultdict(int, FrequencyTable(window, k))
    
    # Check patterns in the first window
    for s in freqMap:
        if freqMap[s] >= t:
            Patterns.add(s)
            
    # Slide the window
    for i in range(1, n - L + 1):
        # Remove the first k-mer of the previous window
        first_pattern = Text[i-1:i-1+k]
        freqMap[first_pattern] -= 1
        
        # Add the last k-mer of the current window
        last_pattern = Text[i+L-k:i+L]
        freqMap[last_pattern] += 1
        
        # Check if the new k-mer meets the threshold
        if freqMap[last_pattern] >= t:
            Patterns.add(last_pattern)
            
    return list(Patterns)
```

```python
genome = open("E_coli.txt", "r").read()
k = 9
L = 500
t = 3
clumps = FindClumps(genome, k, L, t)
print(" ".join(clumps))
print(len(clumps))
```
