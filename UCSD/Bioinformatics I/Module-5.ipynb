**When using motif software with fixed motif lengths, is it better to start with short motifs or long motifs? Why?**

Answer: Short motifs are better to start with because most biologically relevant motifs tend to be short. Additionally, longer motifs tend to be combinations of shorter motifs, so starting with the shorter motifs allows their discovery and subsequent identification much easier.

To evaluate the statistical significance of an identified motif, we need to ensure that a motif with the same or even larger score is unlikely to occur in a collection of "typical" DNA strings (of the same length).

**How would you generate these strings?  Justify your answer.**

Answer: Real DNA is not random, so generating random strings will not provide the desired outcome. The strings should be generated following the typical distribution of DNA in the original DNA strings. Using probabilities of observing each single base, as well as pairs and triples (basically each k-mer), then the generated strings will more closely resemble a collection of "typical" DNA strings.

```python
reference_string = "57 139 107 172 114 136 159 143 155 186 178 200 118 137 173 201 160 62 216 165 45 204"
input_string = "107 139 57 160 201 137 204 173 159 178 186 165 114 172 118 62 200 216 45 155 143 136" # Paste your input string here

def get_missing_integers(ref, inp):
    ref_set = set(map(int, ref.split()))
    if not inp:
        return len(ref_set), sorted(list(ref_set))
    inp_set = set(map(int, inp.split()))
    missing = ref_set - inp_set
    return len(missing), sorted(list(missing))

missing_count, missing_values = get_missing_integers(reference_string, input_string)
print(f"Missing count: {missing_count}")
print(f"Missing values: {missing_values}")
```
